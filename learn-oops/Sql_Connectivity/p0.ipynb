# SQL Connectivity

✅ What is SQLite?

SQLite is a lightweight, file-based database.
👉 No server needed. Your database is just a .db file.

🔹 Method: SQLite Connectivity in Python
Step 1 — Import sqlite3 module
import sqlite3

Step 2 — Create / Connect to Database

If the file doesn’t exist, SQLite will automatically create it.

conn = sqlite3.connect("mydatabase.db")

Step 3 — Create a Cursor

Cursor is used to execute SQL queries.

cursor = conn.cursor()

Step 4 — Create a Table
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER
)
""")

Step 5 — Insert Data
cursor.execute("INSERT INTO students (name, age) VALUES (?, ?)", ("Suyash", 21))
conn.commit()   # Save changes

Step 6 — Fetch Data
cursor.execute("SELECT * FROM students")
rows = cursor.fetchall()

for row in rows:
    print(row)

Step 7 — Close Connection (Important)
conn.close()

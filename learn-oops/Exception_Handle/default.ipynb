## Default Exception Handling

- By the help of Default Exception Handling all exception including KeyboardInterrupt Error.
- Default Exception Handling refers to the automatic way a programming language handles an exception when you do NOT write your own try-except (or try-catch) block.

**In simple words:**

- If you do not handle an exception yourself, the language’s runtime system handles it by default.

## 🔹 What happens in Default Exception Handling?

- When an error occurs and you don’t handle it, this happens automatically:

1. Program execution stops
2. An exception object is created
3. The runtime prints an error message (traceback)
4. The program terminates abnormally (crashes) ❌

## 🔹 How is this different from Custom Exception Handling?

| Default Exception Handling   | Custom Exception Handling      |
| ---------------------------- | ------------------------------ |
| Done automatically by system | Done manually by programmer    |
| Crashes the program          | Prevents crash                 |
| Shows technical error        | Can show user-friendly message |
| No recovery mechanism        | Can recover and continue       |


## 🔹 Why does Default Exception Handling exist?

It ensures that:
- Errors are visible
- Bugs are not silently ignored
- Developers can debug using traceback

In other words, it’s a built-in safety and debugging mechanism.

*Note: Default Exception Handling is the automatic handling of runtime errors by the programming language when the programmer does not provide explicit exception handling.*


*Note: when we know the exact type of exception then use `specific exception` handling otherwise/else we use `Default Exception` Handling.*

```python
# 🔹 Example of Custom Handling (For comparison)
try:
    x = 10 / 0
except ZeroDivisionError:
    print("You cannot divide by zero")
print("Program continues...")
```

```python
try:
    i=0
    while True:
        print("Pirates",i)
        i+=1
except:
    print("Some Error Occurred")
else:
    print("No Error Occurred")
finally:
    print("Execution Completed")
```

## Types of exception 
1. user defined exception
2. inbuilt exception



### User Defined 👍
- By extending/inheriting Exception class we can create our user defined class.
- `raise` keyword is used to throw error according to user.
- by using raise keyword we can raise both user defined exception and inbuilt exception.

**Syntax :**

        `raise ExceptionName("Msg")`






```python
age=int(input("Enter your age: "))
if age<18:
    raise ValueError("Age must be at least 18 to register.")
print("Registration successful.")
```

### User Defined Exception:

- These are exceptions created by the programmer when built-in exceptions are not sufficient.

**Definition:**

- User-defined exceptions are custom exceptions created by inheriting from the base Exception class.


## 🔹 Difference (Short & Simple)

| Feature         | Inbuilt Exception   | User-Defined Exception |
| --------------- | ------------------- | ---------------------- |
| Who creates it? | Language            | Programmer             |
| Example         | `ZeroDivisionError` | `AgeError`             |
| Need to define? | No                  | Yes                    |
| Used for        | Common errors       | Custom logic           |



```python
class AgeError(Exception):
    pass

age = int(input("Enter age: "))

if age < 18:
    raise AgeError("You are not eligible to vote")
else:
    print("You are eligible")
```

```python
class AgeError(Exception):
    pass

try:
    age = int(input("Enter age: "))
    if age < 18:
        raise AgeError("Not eligible")
    print("Eligible")
except AgeError as e:
    print(e)
```

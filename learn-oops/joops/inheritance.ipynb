##  Inheritance (  Tu mera bacha hai ):
- It's a process of deriving the properties from one class to another class.
- the class from which the properties transfer is known as super/parent/base class.
- the class to which the properties are transfer is known as child/subclass/derived class.

        Inheritance

        ┌──────────┐
        |  Parent  |
        └──────────┘
              │
              │
        ┌──────────┐
        |  Child   |
        └──────────┘


**Syntax**

          class Parent:
               pass
          class Child(Parent):
               pass



## Types of Inheritance 👍

### Single Level Inheritance: (One to one)

- Here, properties are transferred from a single parent class to a single child class, but the vice-versa is not possible.

        Single Inheritance

        ┌──────────┐
        |  Parent  |
        └──────────┘
              │
              │
        ┌──────────┐
        |  Child   |
        └──────────┘


**Syntax :**

          class Parent:
               pass
          class Child(Parent):
               pasS




```python
class Demo:
    a=10
    b=20
class New(Demo):
    c = "This is New class"

ob1 = Demo()
ob2 = New()
print(ob1.a)
print(ob1.b)
print(ob2.a)
print(ob2.b)
print(ob2.c)
```

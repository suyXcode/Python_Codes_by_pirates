# 🔐 Encapsulation in Python (Full Explanation)
## 🔹 What is Encapsulation?

**Encapsulation means:**

- Binding data (variables) and methods (functions) together into a single unit (class)
and restricting direct access to some of the object’s data.
- This pillar is used to provides security of your data.
- It restrict direct access to the data 
- 📌 In simple words:
    - **👉 “Data ko protect karna + controlled access dena”**

## 🔹 Why Encapsulation is Needed?

- ✅ Data protection
- ✅ Prevents accidental modification
- ✅ Improves maintainability
- ✅ Follows real-world security model (ATM, Bank Account)


## 🔹 How Encapsulation Works in Python?

- We provide security by the help of access modifier.

- Python achieves encapsulation using:

    - Public members/access modifier 👍
        - It is allowing the data to be accessed to the outside of the class also.
    - Protected members/access modifier (_variable) 👍
        - We use single underscore before the data or method for protected access modifier.
        - These access inside child class and outside of class.
    - Private members/access modifier (__variable)
        - Private members Accessible only inside the same class
        - can not access in child class and outside of the class.
        - We Use double underscore __ before the data or methods to make it private access sepcifier
        - Enforced via name mangling


## 🔹 Access Levels in Python

| Access Type | Syntax  | Meaning                            |
| ----------- | ------- | ---------------------------------- |
| Public      | `var`   | Accessible everywhere              |
| Protected   | `_var`  | Accessible inside class & subclass |
| Private     | `__var` | Accessible only inside class       |




````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # Private variable

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient balance")

    def get_balance(self):
        return self.__balance


# Object creation
acc = BankAccount(1000)

acc.deposit(500)
acc.withdraw(300)

print(acc.get_balance())   # 1200
````

## 1️⃣ Public Access Modifier
- 🔹 Definition

- Accessible from anywhere
- Default access level in Python


**🔹 Syntax**
```python
variable_name
```


````python
class Student:
    def __init__(self):
        self.name = "Suyash"   # Public variable

    def show(self):
        print(self.name)

s = Student()
print(s.name)   # Accessible
s.show()
````

- 📌 No restriction
- 📌 Used when data is safe to expose


---
## 2️⃣ Protected Access Modifier

- 🔹 Definition

- Accessible inside the class and its child classes
- Uses single underscore _
- Just a convention, not strict

**🔹 Syntax**
```python
_variable
```


````python
class Parent:
    def __init__(self):
        self._money = 5000   # Protected variable

class Child(Parent):
    def show(self):
        print(self._money)

c = Child()
c.show()
````

- 📌 Can still be accessed outside (but not recommended)

```python
print(c._money)  # Works, but bad practice
```

---



## 3️⃣ Private Access Modifier
**🔹 Definition**

- Accessible only inside the same class
- Uses double underscore __
- Enforced via name mangling

**🔹 Syntax**
```python
__variable
```

**🔹 Example**

````python
class Account:
    def __init__(self):
        self.__balance = 1000   # Private variable

    def show_balance(self):
        print(self.__balance)
 
a = Account()
a.show_balance()

# print(a.__balance)  ❌ Error
````

## 🔍 Name Mangling (Interview Gold ✨)

```python
print(a._Account__balance)  # Works
```

Python internally changes:
```python
__balance → _Account__balance
```

- ⚠️ Not recommended to use directly.

## 🔥 Comparison Table (Very Important)


| Access Modifier | Symbol  | Accessibility       |
| --------------- | ------- | ------------------- |
| Public          | `var`   | Anywhere            |
| Protected       | `_var`  | Class + Child class |
| Private         | `__var` | Same class only     |

## @property Decorator 👍

- @property is a very important Pythonic concept and interview favorite because it shows encapsulation + clean design.
- @property allows us to use the method as properties(without parenthesis.).
- @property is a built-in decorator that allows you to:

    - Access a method like an attribute (variable) while still having full control over getting, setting, and deleting data.


## 🔹 Why Do We Need @property?

- Without @property:
```python
obj.get_value()
obj.set_value(10)
```

- With @property:
```python
obj.value
obj.value = 10
```

- 👉 Clean
- 👉 Readable
- 👉 Professional


## 🔥 Advantages of @property

- ✔️ Data encapsulation
- ✔️ Validation logic
- ✔️ No need for getter/setter methods
- ✔️ Backward compatibility
- ✔️ Clean API design

## ⚠️ Interview Traps
**🚫 Trap 1: Property is a variable**

- ❌ Wrong
- ✅ It is a method accessed like a variable

## 🚫 Trap 2: Property can exist without a method

- ❌ Wrong
- ✅ Property is always created from a method

## 🚫 Trap 3: Property improves performance
- ❌ Wrong
- ✅ It improves design, not performance

**🎯 Interview Answers**
- 📝 One-Line

**@property allows a method to be accessed like an attribute while enforcing encapsulation.**

- 📝 Two-Line

**The @property decorator is used to define getter, setter, and deleter methods in a clean and Pythonic way. It helps in controlled access to private variables.**

## 🧠 When to Use @property?

- ✅ When validation is needed
- ✅ When logic may change later
- ✅ When public API should remain stable


````python
class Student:
    def __init__(self):
        self.__marks = 0   # Private variable

    @property
    def marks(self):
        return self.__marks
````

````python
s = Student()
print(s.marks)   # Looks like variable access
# 📌 Internally, it calls the marks() method.
````

````python
class Temperature:
    def __init__(self, celsius):
        self.__celsius = celsius

    @property
    def celsius(self):
        return self.__celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            print("Temperature below absolute zero!")
        else:
            self.__celsius = value

    @property
    def fahrenheit(self):
        return (self.__celsius * 9/5) + 32
````

````python
t = Temperature(25)
print(t.fahrenheit)
````

````python
class Person:
    age = 20
    @property
    def height(self):
        return 170
p = Person()
print(p.age)     # 20
print(p.height)  # 170
````

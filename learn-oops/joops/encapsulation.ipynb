# 🔐 Encapsulation in Python (Full Explanation)
## 🔹 What is Encapsulation?

**Encapsulation means:**

- Binding data (variables) and methods (functions) together into a single unit (class)
and restricting direct access to some of the object’s data.
- This pillar is used to provides security of your data.
- It restrict direct access to the data 
- 📌 In simple words:
    - **👉 “Data ko protect karna + controlled access dena”**

## 🔹 Why Encapsulation is Needed?

- ✅ Data protection
- ✅ Prevents accidental modification
- ✅ Improves maintainability
- ✅ Follows real-world security model (ATM, Bank Account)


## 🔹 How Encapsulation Works in Python?

- We provide security by the help of access modifier.

- Python achieves encapsulation using:

    - Public members/access modifier 👍
        - It is allowing the data to be accessed to the outside of the class also.
    - Protected members/access modifier (_variable) 👍
        - We use single underscore before the data or method for protected access modifier.
        - These access inside child class and outside of class.
    - Private members/access modifier (__variable)
        - Private members Accessible only inside the same class
        - can not access in child class and outside of the class.
        - We Use double underscore __ before the data or methods to make it private access sepcifier
        - Enforced via name mangling


## 🔹 Access Levels in Python

| Access Type | Syntax  | Meaning                            |
| ----------- | ------- | ---------------------------------- |
| Public      | `var`   | Accessible everywhere              |
| Protected   | `_var`  | Accessible inside class & subclass |
| Private     | `__var` | Accessible only inside class       |




```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # Private variable

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient balance")

    def get_balance(self):
        return self.__balance


# Object creation
acc = BankAccount(1000)

acc.deposit(500)
acc.withdraw(300)

print(acc.get_balance())   # 1200
```

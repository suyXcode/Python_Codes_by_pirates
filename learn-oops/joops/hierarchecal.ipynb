# Hierarchecal Inheritance : 
---
- In this properties are transfered from single parent class to multiple child class.




                 ┌──────────┐
                 |  Parent  |
                 |    A     |
                 └──────────┘
                 /        \          
                /          \ 
         ┌──────────┐      ┌──────────┐
         |   Child  |      |   Child  |
         |    c     |      |    B     |
         └──────────┘      └──────────┘



**Syntax :**
--- 

    class Pc:
        sb
    class cc1(Pc):
        sb
    class cc2(Pc):
        sb



```python
class Parent:
    s ="This is parent class."

class A(Parent):
    a="hi"
class B(Parent):
    b="hello"
    
ob1 = A()
print(ob1.s)
print(ob1.a)
ob2 = B()
print(ob2.s)
print(ob2.b)
```

```python
# Example of Hierarchical Inheritance

class Parent:
    def show_parent(self):
        print("This is the Parent class.")
class Child1(Parent):
    def show_child1(self):
        print("This is Child1 class inheriting from Parent.")
class Child2(Parent):
    def show_child2(self):
        print("This is Child2 class inheriting from Parent.")
obj1 = Child1()
obj1.show_parent()
obj1.show_child1()
obj2 = Child2()
obj2.show_parent()
obj2.show_child2()
```

# Method Chaining in Python

## What is Method Chaining?
**Method chaining** is a programming technique where **multiple methods are called on the same object in a single line**, one after another.

👉 This is possible **only when each method returns the object itself (`self`)**.

---

## Simple Definition (Interview Ready)
> **Method chaining allows multiple method calls on the same object by returning `self` from each method.**

---

## Why Use Method Chaining?
- ✔ Cleaner and readable code
- ✔ Less repetitive object names
- ✔ Fluent / expressive style
- ✔ Common in APIs, builders, and configuration objects
---
**Syntax**
```python
class ClassName:
    def method1(self):
        .
        .
        return self
    def method2(self):
        .
        .
        return self
        . 
        . 
        .

    def methodn(self):
        .
        .
        return self
        
```
---

## Basic Example

```python
class Calculator:
    def add(self, a, b):
        self.result = a + b
        return self

    def sub(self, c):
        self.result -= c
        return self

    def mul(self, d):
        self.result *= d
        return self

    def show(self):
        print(self.result)
        return self
```

# Method Chaining by jitu sir
- It is a process of calling multiple method in single object in single line of expression.
- In order to achieve method chaining the methods present inside the class should return object reference / self 


````python
class Person:
    def __init__(self):
        self.name = None
        self.age = None
        
    
    def set_name(self, new_name):
        self.name = new_name
        return self
    def set_age(self, new_age):
        self.age = new_age
        return self
    
    def display(self):
        print(f"Name: {self.name}, Age: {self.age}")
        return self
p = Person()
p.set_name("Alice").set_age(30).display()
````

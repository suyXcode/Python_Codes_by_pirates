# 🪄 What is a Magic Method?

## 🔑 Definition

**Magic methods are special methods in Python that:**

- Start and end with double underscores
- Are automatically called by Python
- Allow you to customize class behavior

- 👉 Also called Dunder methods (__method__)

## 🧠 Why Are They Called “Magic”?

- Because you don’t call them directly—Python calls them behind the scenes ✨

**Example:**

```python
print(obj)

**➡ Internally calls:**

obj.__str__()
```




````python
class Collection:
    def __init__(self, items):
        self.items = items
    # Implementing __len__ to return the number of items in the collection
    def __len__(self):
        return len(self.items)
    # Implementing __getitem__ to allow indexing into the collection
    def __getitem__(self, index):
        return self.items[index]
    # Implementing __setitem__ to allow setting values at specific indices in the collection
    def __setitem__(self, index, value):
        self.items[index] = value

coll = Collection([1, 2, 3, 4, 5])
print(len(coll))        # Output: 5
print(coll[2])         # Output: 3
coll[2] = 10
print(coll[2])         # Output: 10
````

# 📌 Commonly Used Magic Methods

| Method        | Purpose                   |
| ------------- | ------------------------- |
| `__init__`    | Constructor               |
| `__str__`     | String representation     |
| `__repr__`    | Developer-friendly output |
| `__len__`     | Length of object          |
| `__add__`     | `+` operator              |
| `__eq__`      | `==` comparison           |
| `__getitem__` | Indexing                  |
| `__call__`    | Call object like function |


## 🧪 Basic Example


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"{self.name} scored {self.marks}"

s = Student("Suyash", 95)
print(s)
````

## ➕ Operator Overloading with Magic Methods
Example: `__add__`

````python
class Box:
    def __init__(self, volume):
        self.volume = volume

    def __add__(self, other):
        return self.volume + other.volume

b1 = Box(10)
b2 = Box(20)

print(b1 + b2)
````

**👉 + is overloaded using __add__**

---

### 🔍 Comparison Magic Methods

````python
class Product:
    def __init__(self, price):
        self.price = price

    def __eq__(self, other):
        return self.price == other.price

p1 = Product(100)
p2 = Product(100)

print(p1 == p2)
````

## 📦 Indexing – `__getitem__`

````python
class Data:
    def __init__(self, items):
        self.items = items

    def __getitem__(self, index):
        return self.items[index]

d = Data([10, 20, 30])
print(d[1])
````

## 📞 Callable Objects – `__call__`

````python
class Greeting:
    def __call__(self, name):
        return f"Hello {name}"

g = Greeting()
print(g("bhai"))
````

### ⚠️ Important Rules

- ✔ Magic methods must have exact names
- ✔ Never invent your own dunder methods
- ✔ Use them only when behavior customization is needed

#### 🧩 Real-Life Analogy

- `__init__` → Birth of object 👶

- `__str__` → How object introduces itself 🗣

- `__add__` → How objects combine 🤝

### 💡 One-line Summary

- Magic methods let your objects behave like built-in Python types

## 🪞 __str__ vs __repr__
- 🔑 One-line Difference

- `__str__` → for users (readable)
- `__repr__` → for developers (unambiguous, debug-friendly)

## 📌 `__str__` — User-Friendly Output

**🔹 What it does**

- Controls output of:

    - print(obj)

    - str(obj)

- Should be clean, readable, informal

✅ Example

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"{self.name} scored {self.marks}"

s = Student("Suyash", 95)
print(s)
````

- ✔ Easy to read
- ✔ End-user friendly

--- 


## 🧠 __repr__ — Developer / Debug Output

**🔹 What it does**

- Controls output of:

    - obj in interpreter
    - repr(obj)

- Should be precise
- Ideally recreate the object



**✅ Example**


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __repr__(self):
        return f"Student(name={self.name!r}, marks={self.marks})"

s = Student("Suyash", 95)
s
````

- ✔ Useful for debugging
- ✔ Clear internal state

##  📊 Comparison Table

| Feature         | `__str__` | `__repr__`        |
| --------------- | --------- | ----------------- |
| Audience        | End user  | Developer         |
| Used by         | `print()` | Interpreter       |
| Output          | Readable  | Precise           |
| Debugging       | ❌         | ✅                 |
| Recreate object | ❌         | ✅ (ideally)       |
| Fallback        | ❌         | Used by `__str__` |


## 🧩 Real-Life Analogy

- `__str__` 🗣
    - 👉 “Hi, I’m Suyash. I scored 95.”

- `__repr__` 🛠
    - 👉 Student(name='Suyash', marks=95)

## 🎯 Interview-Ready Summary

- ✔ `__str__` → Human-readable
- ✔ `__repr__` → Developer-oriented
- ✔ If `__str__` missing → Python uses `__repr__`


````python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} costs ₹{self.price}"

    def __repr__(self):
        return f"Product(name={self.name!r}, price={self.price})"

p = Product("Laptop", 60000)

print(p)   # __str__
p          # __repr__
````

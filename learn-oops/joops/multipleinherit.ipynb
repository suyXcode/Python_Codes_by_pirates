```python
class Math:
    a="This is a Math class"
class Excel:
    b="This is an Excel class"
class SQL:
    c="This is an SQL class"
class DataScience(Math, Excel, SQL):
    pass
```

```python
ob1=DataScience()
print(ob1.a)
print(ob1.b)
print(ob1.c)
```

```python
class Add:
    @staticmethod
    def add(a, b):
        print(a + b)

class Sub:
    @staticmethod
    def sub(a, b):
        print(a - b)

class Pro:
    @staticmethod
    def pro(a, b):
        print(a * b)

class Div:
    @staticmethod
    def div(a, b):
        print(a / b)

class Cal(Add, Sub, Pro, Div):
    pass

c = Cal()
c.add(12, 11)
```

```python

c.add(12,11)
c.sub(12,11)
c.pro(12,11)
c.div(12,11)
```

# 🔹 1. Method Overloading
**✅ What it means**

- Method Overloading =
    - ➡️ Same method name
    - ➡️ Different parameters (number/type)

**❗ Important (Python Reality)**

- Python does NOT support traditional method overloading like Java or C++.
 - But we simulate it using:

    - Default arguments

    - *args / **kwargs



```python
class Math:
    def add(self, a, b=0, c=0):
        return a + b + c

m = Math()
print(m.add(2))        # 2
print(m.add(2, 3))     # 5
print(m.add(2, 3, 4))  # 9
```

**📌 Same method name add() → behaves differently based on arguments.**

# 🔹 2. Method Overriding
**✅ What it means**

- Method Overriding =
    - ➡️ Child class redefines a method of the parent class
    - ➡️ Same method name + same parameters
    - ➡️ Happens in Inheritance

**🧠 Example**

```python
class Parent:
    def show(self):
        print("This is Parent class")

class Child(Parent):
    def show(self):
        print("This is Child class")

obj = Child()
obj.show()
```

**📌 Child method overrides Parent method.**

# 🔹 3. Duck Typing 🦆
**✅ What it means**

    `💬 “If it looks like a duck and quacks like a duck, it must be a duck.”`

- “If it looks like a duck and quacks like a duck, it is a duck.”
- If it walks like a duck and quacks like a duck, it is a duck.
- python always checks the behavoiurs not the class.
- python doesn't care about the class in which the object belongs, it cares that objects has the required method or not.  


- Duck Typing is a concept in Python where
    - 👉 the type of an object is not important
    - 👉 the behavior (methods it supports) is important
    - ➡️ Python doesn’t care about object type
    - ➡️ It cares about method behavior

## 🔹 Why Duck Typing Exists in Python?


- **Python is a dynamically typed language:**

    - No strict type declarations
    - Type is decided at runtime
    - Encourages flexibility & polymorphism

**Duck typing is runtime polymorphism.**

**🧠 Example**

```python
class Dog:
    def speak(self):
        print("Bark")

class Cat:
    def speak(self):
        print("Meow")

def animal_sound(animal):
    animal.speak()

animal_sound(Dog())
animal_sound(Cat())
```

## 🔍 Explanation

- animal_sound() does not care if object is Dog or Cat

- It only checks:
    - ✔️ Does the object have speak() method?



**🔹 Duck Typing WITHOUT Inheritance 🚫**

```python
class Car:
    def move(self):
        print("Car is moving")

class Person:
    def move(self):
        print("Person is walking")

def start(obj):
    obj.move()

start(Car())
start(Person())
```

- 📌 No parent class
- 📌 No common base class
- 📌 Same method name → works perfectly




- **📌 No inheritance needed**
- **📌 Same method name → Python is happy 😄**

## 🔹 Duck Typing vs Traditional Polymorphism

| Feature          | Duck Typing  | Traditional Polymorphism |
| ---------------- | ------------ | ------------------------ |
| Language Type    | Dynamic      | Static                   |
| Inheritance      | Not required | Required                 |
| Type Checking    | Runtime      | Compile time             |
| Flexibility      | Very High    | Limited                  |
| Example Language | Python       | Java, C++                |

## 🔹 Advantages of Duck Typing ✅

- ✔️ Highly flexible
- ✔️ Less code
- ✔️ No inheritance needed
- ✔️ Encourages reusable design

## 🔹 Disadvantages of Duck Typing ❌

- ❌ Runtime errors possible
- ❌ Harder debugging
- ❌ No strict method enforcement

## 🔹 Real-World Example 🌍

```python
class EmailNotifier:
    def send(self):
        print("Email sent")

class SMSNotifier:
    def send(self):
        print("SMS sent")

def notify(service):
    service.send()

notify(EmailNotifier())
notify(SMSNotifier())
```

# 📌 Any object with send() works.


## 🔹 One-Line Interview Answer 🎯

- Duck Typing is a concept where an object’s behavior determines its suitability, not its actual type.

## 🔹 2-Line Interview Answer 📝

- Duck typing allows objects of different classes to be used interchangeably if they implement the same methods. Python checks method existence at runtime rather than class hierarchy.

# **🔹 4. Operator Overloading**

**✅ What it means**

- Operator Overloading =
    - ➡️ Giving new meaning to operators (+, -, *, etc.)
    - ➡️ Done using magic methods (__add__, __sub__, etc.)
    - **Python achieves this using magic methods (also called dunder methods).**
-  It allows operator to work on the user defined object(object of the user defined class .)by using some magic methods/dunder methods.

**🧠 Example**

```python

class Arithmetic:
    def __init__(self, a):
        self.a = a
ob1 = Arithmetic(23)
ob2 = Arithmetic(45)

print(ob1.a)  # 23
print(ob2.a)  # 45
print(ob1.a + ob2.a)  # 68
print(type(ob1))  # <class '__main__.Arithmetic'>
print(type(ob2))  # <class '__main__.Arithmetic'>
print(type(ob1.a))  # <class 'int'>
print(type(ob2.a))  # <class 'int'>
# print(ob1 + ob2 )  # TypeError: unsupported operand type(s) for +: 'Arithmetic' and 'Arithmetic'
```

```python
magic_methods = [m for m in dir(int) if m.startswith("__")]
print(magic_methods)
```

```python
class Point:
    def __init__(self, x):
        self.x = x

    def __add__(self, other):
        return self.x + other.x

p1 = Point(10)
p2 = Point(20)

print(p1 + p2)
```

**📌 + operator overloaded using `__add__()`**

## 🎯 Interview One-Line Answers

- Method Overloading: Same method name with different parameters (simulated in Python).

- Method Overriding: Child class provides its own implementation of a parent method.

- Duck Typing: Object behavior matters, not its type.

- Operator Overloading: Operators behave differently using magic methods.

# 🔥 Method Overloading vs Method Overriding
| Basis                     | **Method Overloading**                         | **Method Overriding**                                            |
| ------------------------- | ---------------------------------------------- | ---------------------------------------------------------------- |
| **Definition**            | Same method name with **different parameters** | Child class provides its **own implementation** of parent method |
| **OOP Concept**           | Compile-time polymorphism                      | Runtime polymorphism                                             |
| **Inheritance Required?** | ❌ No                                           | ✅ Yes                                                            |
| **Method Name**           | Same                                           | Same                                                             |
| **Parameters**            | Must be **different**                          | Must be **same**                                                 |
| **Return Type**           | Can be different                               | Should be same                                                   |
| **Binding Time**          | Compile time                                   | Runtime                                                          |
| **Python Support**        | ❌ Not directly (simulated)                     | ✅ Fully supported                                                |
| **Used For**              | Different ways to call same logic              | Change parent behavior                                           |



## 🎯 Key Interview Differences (Short)

- Overloading → Same class, different parameters
- Overriding → Different classes, same parameters
- Overloading → Compile time
- Overriding → Runtime
- Python → Overloading is simulated, overriding is real


## 📝 One-Line Interview Answer

- Method Overloading allows multiple methods with the same name but different parameters, while Method Overriding allows a child class to redefine a method of its parent class with the same signature.

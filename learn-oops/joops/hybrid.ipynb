# Hybrid Inheritance

- It is combination of two or more than types of inheritance.


**Diagram :**

                 ┌──────────┐
                 |  Parent  |
                 |    A     |
                 └──────────┘
                 /        \          
                /          \ 
         ┌──────────┐      ┌──────────┐
         |   Child  |      |   Child  |
         |    C     |      |    B     |
         └──────────┘      └──────────┘
                \            /
                 \          /
                  \        /
                 ┌──────────┐
                 |  Child   |
                 |    D     |
                 └──────────┘

```python
class Math:
    a="This is a Math class"
class Excel:
    b="This is an Excel class"
class SQL:
    c="This is an SQL class"
class DataScience(Math, Excel, SQL):
    pass
class ML(DataScience):
    pass
class AI(DataScience):
    pass

ob1=AI()
print(ob1.a)
print(ob1.b)
print(ob1.c)
```

# constructor chaining 👍
---

- It is process of calling the parent constructor inside the child class.
- If we want to add some extra properties for child class object then we should use constructor chaining .
- we can do so by using `super()` or `class_name`
- constructor call should be done  inside the constructor of the child class.


**Syntax:**

    class Parent:
        def __init__(self):
            sb
    class Child(Parent):
        def __init__(self):
            super().__init__()
            #Parent.__init__(self)



```python
# Constructor chaining example
class Demo:
    def __init__(self):
        print("This is the demo class constructor.")
class Primary(Demo):
    def __init__(self):
        super().__init__()  # Call the constructor of the parent class
        print("This is the Primary class constructor.")
        Demo.__init__(self)  # Call the constructor of the Demo class explicitly
obj = Primary()  # This will call the constructor of both Demo and Primary classes

print("Object created successfully.")
```

```python
class A:
    def __init__(self):
        print("This is class A constructor.")
class B:
    def __init__(self):
        print("This is class B constructor.")
class C:
    def __init__(self):
        print("This is class C constructor.")
class Main(A, B, C):
    def __init__(self):
        C.__init__(self)  # Call the constructor of the parent classes
        print("This is main class constructor.")
obj = Main()  # This will call the constructor of all classes in the order of inheritance
print("Object created successfully.")
```

```python
#by using super()
class A:
    def __init__(self):
        print("This is class A constructor.")
class B:
    def __init__(self):
        print("This is class B constructor.")
class C:
    def __init__(self):
        print("This is class C constructor.")
class Main(A, B, C):
    def __init__(self):
        super().__init__()  # Call the constructor of the first parent class in MRO
        print("This is main class constructor.")
obj = Main()  # This will call the constructor of all classes in the order of inheritance
print("Object created successfully.")
```

```python
# by using direct parent class name
class A:
    def __init__(self):
        print("This is class A constructor.")
class B:
    def __init__(self):
        print("This is class B constructor.")
class C:
    def __init__(self):
        print("This is class C constructor.")
class Main(A, B, C):
    def __init__(self):
        B.__init__(self)  
        C.__init__(self)  
        print("This is main class constructor.")
obj = Main()  
print("Object created successfully.")
```

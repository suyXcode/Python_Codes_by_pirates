# Multiple  Inheritance

- Multiple Inheritance is a type of inheritance in which a child class inherits properties and methods from more than one parent class.

**In simple words:**

**👉 One Child Class ← Multiple Parent Classes**

Syntax :

Class A
Class B
   ↓
Class C (inherits from A and B)


**Diagram**

         ┌──────────┐      ┌──────────┐
         |  Parent  |      |  Parent  |
         |    A     |      |    B     |
         └──────────┘      └──────────┘
                \            /
                 \          /
                  \        /
                 ┌──────────┐
                 |  Child   |
                 |    C     |
                 └──────────┘


```
class A:
    pass

class B:
    pass

class C(A, B):   # C inherits from both A and B
    pass
```

````python
class Addition:
    @staticmethod
    def add(x, y):
        return x + y
class Subtraction:
    @staticmethod
    def subtract(x, y):
        return x - y
class Product:
    @staticmethod
    def multiply(x, y):
        return x * y
class Division:
    @staticmethod
    def divide(x, y):
        if y != 0:
            return x / y
        else:
            return "Error! Division by zero."
class Calculator(Addition, Subtraction, Product, Division):
    pass

ob1 = Calculator()
print("Addition:", ob1.add(10, 5))
print("Subtraction:", ob1.subtract(10, 5))
print("Multiplication:", ob1.multiply(10, 5))
print("Division:", ob1.divide(10, 5))
````

# 📘 Method Resolution Order (MRO)

- Method Resolution Order (MRO) is the order in which Python searches for a method or attribute in inheritance hierarchy when a method is called.
- It searches inside the child class first.
- It searches the parent class in a consistent manner from left to the right.



**In simple words:**

- 👉 MRO tells Python: “Which class to check first, second, third… when calling a method?”

**Python follows the C3 Linearization Algorithm to decide this order.**


## 🔹 Why MRO is Important?
- ✔ Works correctly in multiple inheritance

- When a class has two or more parent classes, MRO avoids confusion about which parent class method should be called first.

- ✔ Helps super() work properly

- super() follows the MRO sequence to call methods in the correct order.

- ✔ Avoids duplicate calls

- Prevents the same class from being called again and again (Diamond Problem).

**📊 Example of MRO**

class A:
    pass

class B(A):
    pass

class C(B):
    pass

print(C.mro())

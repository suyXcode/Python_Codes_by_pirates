# Polymorphism in Python

## What is Polymorphism?
**Polymorphism** means **“many forms”**.

👉 In programming, it refers to the **same method name behaving differently** depending on:
- the **object**
- the **class**
- or the **context**

**Polymorphism is a core concept in Object-Oriented Programming (OOP). The word means "many forms" — and in programming, it allows the same interface or method name to behave differently depending on the object or context.**

---

## Simple Definition (Interview Ready)
> **Polymorphism allows the same method or function to perform different tasks based on the object or data it is working with.**

---

## Why Polymorphism is Used?
- ✔ Improves code flexibility
- ✔ Supports scalability
- ✔ Reduces tight coupling
- ✔ Enables clean OOP design

---

## Types of Polymorphism in Python

1. **Function Polymorphism**
2. **Operator Overloading**
3. **Method Overriding (Runtime Polymorphism)**
4. **Duck Typing**

---

## 1️⃣ Function Polymorphism

Same function name, different behavior.

```python

print(len("Python"))     # String
print(len([1, 2, 3]))    # List

```
**✔ len() works differently for different data types.**

````python
class Animal:
    def speak(self):
        print("Subclasses must implement this method")
class Dog(Animal):
    def speak(self):
        print("Woof!")
class Cat(Animal):
    def speak(self):
        print("Meow!")
dog = Dog()
cat = Cat()
dog.speak()  # Output: Woof!
cat.speak()  # Output: Meow!
````

# types of relation between class 👍

- IS-A (Inheritance)
- HAS-A (Association)


## 2️⃣ HAS-A Relationship (Association)


- It is a type of relation where object of one class will be the property of another class.
- One class contains another class in HAS-A relation.


**✅ Meaning**

One class has an object of another class

Objects are independent

**🔹 Example**


```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self, engine):
        self.engine = engine   # HAS-A relationship

    def drive(self):
        self.engine.start()
        print("Car is moving")

e = Engine()
c = Car(e)
c.drive()
```

- 📌 Car has an Engine
- 📌 Loose coupling

- **Has-A is two type of relation**
    - Aggregation (Weak Has-A)
        - object is independent
        - loosely coupled
        - example : team-player,teacher-student
    - Uses-A Relationship (Dependency)
        - object is dependent on class
        - tightly coupled
        - Ex : house-room,car-enginge

## 🔥 Comparison Table (VERY IMPORTANT)
| Relationship | Keyword     | Strength    |
| ------------ | ----------- | ----------- |
| IS-A         | Inheritance | Strong      |
| HAS-A        | Association | Medium      |
| Aggregation  | HAS-A       | Weak        |
| Composition  | HAS-A       | Very Strong |
| Uses-A       | Dependency  | Very Weak   |


## 🎯 Interview One-Line Answers

- IS-A: One class inherits another class
- HAS-A: One class contains an object of another class
- Aggregation: Weak HAS-A relationship
- Composition: Strong HAS-A relationship
- Uses-A: Temporary usage of another class

## ⚠️ Interview Traps

**❌ Aggregation and Composition are same**
**❌ HAS-A always means Composition**

- ✅ Both are WRONG

## 🧠 Quick Memory Trick

- Inheritance → IS-A
- Car has Engine → HAS-A
- School has Teacher → Aggregation
- Human has Heart → Composition

```python
class Address:
    def __init__(self, city, state):
        self.city = city
        self.state = state
addr = Address("New York", "NY")
class Person:
    def __init__(self, name, address):
        self.name = name
        self.address = address   # HAS-A relationship
p = Person("Alice", addr) # addr is an object of Address classu
print(p.name)
print(p.address.city, p.address.state)  # Alice New York NY
```

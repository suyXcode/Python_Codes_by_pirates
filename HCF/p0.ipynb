# Euclidean Algorithm (Most Important ⭐)
- ✅ Fast and standard method
- 🔥 Must-know for interviews

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

print(gcd(17,10))
```

# Recursive Euclidean Algorithm
- 📌 Clean and elegant recursion



```python
def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)

print(gcd(12, 18))
```

# Using Python Built-in Function

```python
import math

print(math.gcd(12, 18))
```

# Lambda Function Style

```python
gcd = lambda a, b: a if b == 0 else gcd(b, a % b)

print(gcd(12, 18))
```

# 🔁 GCD using Repeated Subtraction (While Loop)

**Logic:**
Subtract the smaller number from the larger one until both become equal.

```python
def gcd(a, b):
    while a != b:
        if a > b:
            a = a - b
        else:
            b = b - a
    return a

print(gcd(12, 18))
```

## 🔁 Using Recursion (Repeated Subtraction)

```python
def gcd(a, b):
    if a == b:
        return a
    if a > b:
        return gcd(a - b, b)
    return gcd(a, b - a)

print(gcd(12, 18))
```

```python
# nth fibbonacci number :- 
def fib(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return fib(n-1)+fib(n-2)
print(fib(10))
```

```python
# tribbonacci number :-
def trib(n):
    if n==0:
        return 0
    elif n==1 or n==2:
        return 1
    else:
        return trib(n-1)+trib(n-2)+trib(n-3)
print(trib(10))
```

```python
# series :-
def ser(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    elif n==2:
        return 2
    else:
        return ser(n-1)+ser(n-2)+ser(n-3)
print(ser(10))
```

```python
# prime number :-
def isprime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
print(isprime(17))
```

```python
# prime number using recursive function  :-
def isprime(n,i=2):
    if n <= 1:
        return False
    if n==i:
        return True
    elif n%i==0:
        return False
    return isprime(n,i+1)
print(isprime(15))
```

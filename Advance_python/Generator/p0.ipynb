# 1️⃣ What is a Generator?

- A generator in Python is a special type of function or object that produces values one at a time, instead of storing all values in memory at once.

- 👉 It uses the yield keyword instead of return.

**Key idea:**

- Generates values on the fly
- Saves memory
- Useful for large data or infinite sequences



**GENERATOR**

- Simple way of creating iterator.
- It is special type of function that produces value at a time by using `yield` insteated of returning all the values at once.
- It pauses execution remember it's state and resumes when next value requested.

# 2️⃣ Generator Function

- A function becomes a generator when it uses yield.

**Example:**
```
def count_up(n):
    i = 1
    while i <= n:
        yield i
        i += 1
```

**Using the generator:**
```
g = count_up(5)

for value in g:
    print(value)

```


# 3️⃣ Generator Expression

- A generator expression is a short, one-line way to create a generator.

- 👉 It looks like a list comprehension, but uses parentheses () instead of [].

**Syntax:**
```
(expression for item in iterable)
```

**Example:**
```
gen = (x * x for x in range(5))
for val in gen:
    print(val)

```


# 4️⃣ Generator vs List

| Feature     | Generator           | List           |
| ----------- | ------------------- | -------------- |
| Memory      | Low (lazy)          | High           |
| Execution   | One value at a time | All at once    |
| Syntax      | `()`                | `[]`           |
| Reusability | Only once           | Multiple times |

## What is yield Keyword?

- 👉 yield is used in a function to make it a generator.
Instead of returning a value and stopping execution, yield returns a value and pauses the function, saving its state.
- yield is used to produce values one at a time from a generator while preserving the function’s state.

### How yield Works (Step by Step)

- Function runs until it hits yield

- Value is returned

- Function pauses

- On next call (next()), execution resumes from the same place

````python
# Example of generator

def demo():
    yield "First"
    yield "Second"
    yield "Third"
print(demo(),type(demo()))  # generator object
````

````python
val = demo()
print(next(val))
print(next(val))
````

````python
print(next(val))
````

````python
val2 = demo()
for item in val2:
    print(item)
````

````python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
fib_gen = fibonacci(10)
for num in fib_gen:
    print(num)
````

# 1️⃣ What is a Generator?

- A generator in Python is a special type of function or object that produces values one at a time, instead of storing all values in memory at once.

- 👉 It uses the yield keyword instead of return.

**Key idea:**

- Generates values on the fly
- Saves memory
- Useful for large data or infinite sequences



**GENERATOR**

- Simple way of creating iterator.
- It is special type of function that produces value at a time by using `yield` insteated of returning all the values at once.
- It pauses execution remember it's state and resumes when next value requested.

# 2️⃣ Generator Function

- A function becomes a generator when it uses yield.

**Example:**
```
def count_up(n):
    i = 1
    while i <= n:
        yield i
        i += 1
```

**Using the generator:**
```
g = count_up(5)

for value in g:
    print(value)

```


# 3️⃣ Generator Expression

- A generator expression is a short, one-line way to create a generator.

- 👉 It looks like a list comprehension, but uses parentheses () instead of [].

**Syntax:**
```
(expression for item in iterable)
```

**Example:**
```
gen = (x * x for x in range(5))
for val in gen:
    print(val)

```

## What is yield Keyword?

- 👉 yield is used in a function to make it a generator.
Instead of returning a value and stopping execution, yield returns a value and pauses the function, saving its state.
- yield is used to produce values one at a time from a generator while preserving the function’s state.

### How yield Works (Step by Step)

- Function runs until it hits yield

- Value is returned

- Function pauses

- On next call (next()), execution resumes from the same place

````python
# Example of generator

def demo():
    yield "First"
    yield "Second"
    yield "Third"
print(demo(),type(demo()))  # generator object
````

````python
val = demo()
print(next(val))
print(next(val))
````

````python
print(next(val))
````

````python
val2 = demo()
for item in val2:
    print(item)
````

````python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
fib_gen = fibonacci(10)
for num in fib_gen:
    print(num)
````

# 🔁 Generator vs Function (Python)

Understanding **generators**, **iterators**, and how they differ from normal functions is 🔑 for writing memory-efficient and high-performance Python code 🚀

---

## 🧠 Generator vs Function

| Feature | Function | Generator |
|------|--------|-----------|
| Keyword used | `return` | `yield` |
| Execution | Executes fully at once | Executes step-by-step |
| Memory usage | Stores full result in memory | Uses **lazy evaluation** (very memory efficient) |
| Output | Single value | Sequence of values (one at a time) |
| State | Does **not** remember state | **Remembers state** between calls |

## 🎯 What Is the Use of `yield` Keyword?

`yield`:

- Returns a value **without stopping** the function  
- Pauses execution and **saves state**  
- On next call, execution resumes from where it paused  

📌 Used to produce values **one at a time**

```python
def count_up():
    for i in range(3):
        yield i
```

---

## 🔄 Difference Between `yield` and `return`

| Feature | return | yield |
|------|--------|-------|
| Stops function | ✅ Yes | ❌ No (pauses only) |
| Multiple values | ❌ No | ✅ Yes |
| State preserved | ❌ No | ✅ Yes |
| Used in | Normal functions | Generator functions |

```python
# return
def func():
    return 10
    return 20  # ❌ unreachable

# yield
def gen():
    yield 10
    yield 20  # ✅ works
```

## 🔁 What Is an Iterator?

An **iterator** is an object that:

- Iterates over elements one at a time  
- Implements two methods:
  - `__iter__()`
  - `__next__()`

```python
nums = iter([1, 2, 3])
print(next(nums))  # 1
print(next(nums))  # 2
```

---

## 🔍 Iterator vs Iterable

| Feature | Iterable | Iterator |
|------|----------|----------|
| Can be looped | ✅ Yes | ✅ Yes |
| Has `__iter__()` | ✅ Yes | ✅ Yes |
| Has `__next__()` | ❌ No | ✅ Yes |
| Examples | list, tuple, string | generator, iterator object |

📌 **Iterable → gives an iterator**  
📌 **Iterator → gives values**

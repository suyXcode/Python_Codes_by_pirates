# ✅ 1. Write Student Records into CSV with "w" mode.

```python
from csv import *
with  open('student.csv','w',newline='') as f:
    var=writer(f)
    #header
    var.writerow(["rn", "name", "age", "class", "percentage"])
    #records
    var.writerow([1, "sachin", 20, "bsc", 80])
    var.writerow([2, "sachin", 20, "bsc", 80])
    var.writerow([3, "sachin", 20, "bsc", 80])
    var.writerow([4, "sachin", 20, "bsc", 80])
    var.writerow([5, "sachin", 20, "bsc", 80])
print("Initial record written sucessfully!")

```

# ✅ 2. Append More Student Records

```python
with open('student.csv','a') as f:
    var = writer(f)
    var.writerow([6, "sachin", 20, "bsc", 80])
    var.writerow([7, "sachin", 20, "bsc", 80])
    var.writerow([8, "sachin", 20, "bsc", 80])
    var.writerow([9, "sachin", 20, "bsc", 80])
    var.writerow([10, "sachin", 20, "bsc", 80])
print("New record appended sucessfully!")    
```

# ✅ 3. Read Records & Calculate Overall Percentage .

```python
total = 0
count = 0

with open("student.csv", "r") as file:
    reader = DictReader(file)
    
    for row in reader:
        total += float(row["percentage"])
        count += 1

overall_percentage = total / count
print("Overall Percentage of all students:", overall_percentage)
```

# ✅ 4. Copy Data to Another CSV File with 5% Extra

```python
with open("student.csv", "r") as file1, open("students_updated.csv", "w", newline="") as file2:
    reader = DictReader(file1)
    fieldnames = reader.fieldnames
    
    writer = DictWriter(file2, fieldnames=fieldnames)
    writer.writeheader()
    
    for row in reader:
        row["percentage"] = min(float(row["percentage"]) + 5, 100)
        writer.writerow(row)

print("Data copied with 5% extra marks")
```

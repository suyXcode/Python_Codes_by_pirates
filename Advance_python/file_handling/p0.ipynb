## 📂 What is File Handling in Programming?

- File handling is the mechanism that allows a program to create, read, write, update, and delete files stored on a disk (like HDD/SSD). It enables programs to persist data even after execution ends—unlike variables that exist only in memory (RAM). 🚀
- Reading,Writting and manage the file is known as file handling.
- To perform file handling we have to get the access of file which can be done with the help of `open()`.

---

## 🔑 Why File Handling Is Important?

    - 💾 Permanent data storage (logs, reports, user data)
    - 🔄 Data sharing between programs
    - 🧠 Memory efficiency (large data handled via files)
    - 📝 Record keeping (audit logs, configurations)

## 📁 Common File Operations

| Operation  | Description               |
| ---------- | ------------------------- |
| **Create** | Make a new file           |
| **Open**   | Access an existing file   |
| **Read**   | Retrieve data from a file |
| **Write**  | Add or overwrite data     |
| **Append** | Add data at the end       |
| **Close**  | Release file resources    |
| **Delete** | Remove a file             |

## 📌 File Modes Explained

- "r" → Read

- "a" → Append
- "x" → Create new file

## 🧠 File Handling Functions (Python)
**1️⃣ open() function**
- Used to open a file.

```python
file = open("filename/path", "mode")
```

**MODE**
- Mode is type of operation we want to perform on file.
- there are three type of basic mode in file handling
    - write mode("w")
    - read mode("r")
    - append ("a")


## Write mode 👍

- It is denote by `w` .
- it is used to write the data inside the file
- In the write mode if the file is doesn't exist then create a new file ,if the file is exist then overwrite the data.
- we used two function to write the data 
    - write()  -> single data
    - writelines() -> multiple data

**write() function**

- Writes data to a file (overwrites in w mode).
```python
f = open("data.txt", "w")
f.write("Hello World")
```


**writelines() function**

- Writes multiple lines using a list.

```python
lines = ["Hello\n", "Python\n"]
f.writelines(lines)
```

## **syntax**
1.

```python
var= open("file.exe/path",'w')
var.write(data)
var.writelines([data1,data2,data3,....])
var.close()
```

2. 

```python
with open("file.exe/path",'w') as var:
    var.write(data)
    var.writelines([data1,data2,data3,....])

```

## append mode

- it is denote by `a`.
- It is used to update the file 
- it is the same as the write but it doesn't overwrite the file

````python
var = open("suy.txt",'a') # read is default mode
d1 = "hello world\n"
d2 = "welcome to python programming\n"
d3 = "file handling in python\n"
var.write(d1)
var.writelines([d2,d3])
# var.flush()  # Ensure data is written to the file
var.close()
````

## read mode 👍
- It is denoted by "r".
- It is used to read the data file.
- in read mode if file is not present then it throws error.
- It have three function
    - read()
    - readline()
    - readlines()


### read() function

- IT Reads entire/total content of the file.
- It returns a string.
```python
f = open("data.txt", "r")
print(f.read())
```

### read(n) function

- Reads n characters from the file.

```python
print(f.read(5))   # reads first 5 characters
```

### readline() function

- Reads one line at a time.
- It reads file line by line.

```python
print(f.readline())

```

### readlines() function

- Reads all lines and returns a list.

```python
print(f.readlines())
```




````python
with open("suy.txt") as var:
    data = var.readline()
    data1 = var.readline()
    print(data)
    print(data1)
````

## ✍️ Writing to a CSV File

### Using csv.writer()
- 📌 newline="" avoids blank lines (important!)




````python
import csv

with open("data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Roll", "Name", "Marks"])
    writer.writerow([1, "Amit", 85])
    writer.writerow([2, "Suyash", 92])
    writer.writerow([3, "Ravi", 78])
````

### Writing Multiple Rows – writerows()

````python
with open("data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Roll", "Name", "Marks"])
    writer.writerow([1, "Amit", 85])
    writer.writerow([2, "Suyash", 92])
    rows = [
    [1, "Amit", 85],
    [2, "Suyash", 92]
    ]

    writer.writerows(rows)
````

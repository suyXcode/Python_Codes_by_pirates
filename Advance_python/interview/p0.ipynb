```python
def groupAnagrams(words):
    mp = {}   # hash map

    for word in words:
        # Step 1: count characters manually
        count = [0] * 26   # for a–z

        for ch in word:
            index = ord(ch) - ord('a')
            count[index] += 1

        # Step 2: convert list to tuple (hashable)
        key = tuple(count)

        # Step 3: group words
        if key not in mp:
            mp[key] = []
        mp[key].append(word)

    # Step 4: collect result
    result = []
    for val in mp.values():
        result.append(val)

    return result
```

```python
print(groupAnagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))
```

```python
#wap two check two strings are anagram or not
def areAnagrams(str1, str2):
    if len(str1) != len(str2):
        return False
    count = [0] * 26  # for a–z
    for ch in str1:
        index =ord(ch)-ord('a')
        count[index]+=1
    for ch in str2:
        index =ord(ch)-ord('a')
        count[index]-=1 
    for c in count:
        if c != 0:
            return False
    return True
print(areAnagrams("listen", "silent"))  # True
print(areAnagrams("hello", "world"))    # False
```

```python
def groupbychar(strs):
	groups={}
	for s in strs:
		key=tuple(sorted(s))
		if key not in groups:
			groups[key] = []
		groups[key].append(s)
	print(list(groups.values()))
groupbychar(["eat", "tea", "tan", "ate", "nat", "bat"])
```

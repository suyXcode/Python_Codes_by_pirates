# Package Architecture : 

- It's a process of dividing project into number of folders
- Package Architecture is the structured way of organizing packages, modules, and classes in a software system so that the code is:

    - ✅ Well-organized
    - ✅ Maintainable
    - ✅ Scalable
    - ✅ Reusable

- 👉 It defines how packages are arranged, how they interact, and their responsibilities.


## 🔹 Why Package Architecture is Needed

- 🔧 Avoids messy code
- 🔁 Improves reusability
- 🔒 Controls access & security
- 👥 Helps team collaboration
- 🚀 Makes large systems scalable

## 🔹 Basic Structure (General)
```bash
Project
│
├── ui / presentation
│
├── service / business
│
├── model / entity
│
├── repository / dao
│
└── util / helper
```

**package vs module:**

### 🔹 Basic Definition
**✅ Module**

- A module is a single file that contains Python code (functions, classes, variables).

**✅ Package**

- A package is a collection of modules organized inside a directory.

- Package is a folder which contains n number of files/module where in module contains n numbers of instruction.
- To use the package we install the package where in module we have to import it.
- ex- pip install package_name  (For package) ex - numpy,pandas,fastapi
- ex import module_name (for module) ex - os,sys,turtle,math




## 📊 Package vs Module (Tabular Difference)

| Basis         | Module                        | Package                      |
| ------------- | ----------------------------- | ---------------------------- |
| Definition    | Single Python file            | Directory containing modules |
| File type     | `.py` file                    | Folder                       |
| Purpose       | Organize related code         | Organize multiple modules    |
| Contains      | Functions, classes, variables | Modules + sub-packages       |
| `__init__.py` | ❌ Not required                | ✅ Required (traditionally)   |
| Size          | Small                         | Large                        |
| Reusability   | Limited                       | High                         |
| Example       | `math_utils.py`               | `utils/`                     |





## Module : 

- there are two types of module.
- A module is a single Python file (.py) that contains functions, classes, and variables, which can be imported and reused in other programs.

**🔹 Why Modules are Used?**

- ♻️ Code reusability

- 📦 Better organization

- 🔒 Avoid name conflicts

- 🧠 Easy maintenance

- 👥 Team collaboration


### 🔹 How Many Ways to Import a Module in Python?

- 👉 There are 5 main ways to import a module in Python ✅
---
✅ Advantage

- Clear and safe (no name conflict)

---

## 1️⃣ import module_name


````python

import math
print(math.sqrt(16))
````

## 2️⃣ import module_name as alias
- ✅ Advantage

Short and readable

````python
import numpy as np
print(np.array([1, 2, 3]))
````

## 3️⃣ from module_name import specific_item

⚠ Disadvantage

- Possible name conflict

````python
from math import sqrt
print(sqrt(25))
````

## 4️⃣ from module_name import *
- ❌ Not Recommended

- Pollutes namespace

- Hard to debug

````python
from math import *
print(sin(0))
````

## 📊 Summary Table

| No. | Import Method                | Example                 |
| --- | ---------------------------- | ----------------------- |
| 1   | `import module`              | `import math`           |
| 2   | `import module as alias`     | `import numpy as np`    |
| 3   | `from module import name`    | `from math import sqrt` |
| 4   | `from module import *`       | `from math import *`    |
| 5   | Custom / user-defined import | `from utils import add` |

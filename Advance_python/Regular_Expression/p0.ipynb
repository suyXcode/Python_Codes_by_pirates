# Regular Expression : 

## 🔍 What is a Regular Expression (Regex) in Python?

- A Regular Expression (often called regex) is a powerful pattern-matching technique used to search, match, extract, or replace text in strings.
- In Python, regex is handled using the built-in re module 🚀.
- It is a concept by the help of which can search and match a pattern in a string.
- It is a concept by the help of which can search and match a pattern in a string by using some expression is known as regular expression.


**Syntax**
```python
re.method_name("expression",string)
```


## 🧠 Why Use Regular Expressions?

Regex helps you:

- 🔎 Search for specific patterns in text

- ✂️ Extract useful data (emails, phone numbers, IDs)

- 🔁 Replace or validate strings

- 🧪 Validate inputs (passwords, usernames, formats)



## 📦 Python Regex Module

- Python provides the re module for regex operations:

```python
import re
```

## 🔧 Common Regex Functions in Python

| Function       | Description                             |
| -------------- | --------------------------------------- |
| `re.match()`   | Matches pattern at **start** of string  |
| `re.search()`  | Searches pattern **anywhere** in string  first match|
| `re.findall()` | Returns **all matches**                 |
| `re.sub()`     | Replaces pattern with new text          |
| `re.split()`   | Splits string by pattern                |
| `re.fullmatch()` | Match entire string    |
| `re.finditer()`  | Iterator with position |
| `re.subn()`      | Replace + count        |
| `re.compile()`   | Pre-compile regex      |
| `re.escape()`    | Escape special chars   |



````python
import re

text = "I love Python"
result = re.search("Python", text)

print(result.group())
````

## 🔤 Common Regex Patterns

| Pattern  | Meaning                         |
| -------- | ------------------------------- |
| `.`      | Any single character  |
| `^`      |  It matches at Start of string                 |
| `$`      | It matches at End of string                   |
| `*`      | 0 or more occurrences           |
| `+`      | 1 or more occurrences           |
| `?`      | 0 or 1 occurrence               |
| `\d`     | Digit (0–9)                     |
| `\w`     | Alphanumeric (a–z, A–Z, 0–9, _) |
| `\s`     | Whitespace                      |
| `[abc]`  | Any one of a, b, c              |
| `[^abc]` | NOT a, b, or c                  |



## 🧠 Key Points to Remember

- Regex is case-sensitive by default
- Use raw strings (r"") to avoid escape issues
- Extremely useful in data cleaning, validation, and automation

````python
# 📌 Example: Find All Numbers
import re

text = "My roll number is 123 and code is 456"
numbers = re.findall(r"\d+", text)

print(numbers)
````

````python
# 🔄 Example: Replace Text
import re

text = "Hello World"
new_text = re.sub("World", "Python", text)

print(new_text)
````

````python
# ✅ Example: Validate Email
import re

email = "test@gmail.com"
pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"

if re.match(pattern, email):
    print("Valid Email")
else:
    print("Invalid Email")
````

## 🔹 1. . → Any Single Character

- 📌 Matches exactly ONE character (except newline \n)
- 🧠 Think of . as: “anything here”

Example

````python
print(re.findall("h.t", "hat hot hit hut"))
````

## 🔹 2. ^ → Start of String

- 📌 Ensures the pattern starts at the beginning
- 🧠 Think of ^ as: start anchor

````python
print(re.findall("^Hello", "Hello World"))
````

-✅ Matches only if string starts with "Hello"

❌ Won’t match:
```
"Say Hello"


```

## 🔹 3. $ → End of String

- 📌 Ensures the pattern ends at the end
- 🧠 Think of $ as: end anchor

````python
print(re.findall("World$", "Hello World"))
````

## 🔹 4. * → 0 or More Occurrences

- 📌 Matches zero, one, or many times
- 🧠 b* → b may appear or may not

````python
print(re.findall("ab*", "a ab abb abbb"))
````

## 🔹 5. + → 1 or More Occurrences

- 📌 Must appear at least once
- 🧠 + means mandatory repetition

````python
print(re.findall("ab+", "a ab abb abbb"))
#❌ "a" not included
````

## 🔹 6. ? → 0 or 1 Occurrence

- 📌 Optional — appears once or not at all
- 🧠 Perfect for optional characters

Example


````python
print(re.findall("colou?r", "color colour"))
````

## 🔹 7. \d → Digit (0–9)

- 📌 Matches any numeric digit
- 🧠 Same as [0-9]

````python
print(re.findall("\d", "Order 123, Code 45"))
````

**✔️ Use a raw string (r"")**

````python
print(re.findall(r"\d", "Order 123, Code 45"))
````

**⚠️ Alternative (NOT recommended)**

This works because:

- \\ → Python converts to \

- Regex then sees \d

But ❌ this is ugly and harder to read, especially in complex patterns.

````python
print(re.findall("\\d", "Order 123, Code 45"))
````

## 🔢 {n} in Regular Expression (Regex)
- Exactly n times
- - Matches the preceding character or group exactly n times

- 📌 It is called a quantifier

**🔍 Example 1: Exact Digits**

````python
text = "PIN: 1234"
print(re.findall(r"\d{4}", text))
````

**🧠 \d{4} → exactly 4 digits**

---

**Example 2: Repeating Characters**

````python
print(re.findall(r"a{3}", "aa aaaa aaa a"))
````

**🧠 Matches only 3 a’s together**

--- 

**🔹 Example 3: Fixed-Length Word**


````python
print(re.findall(r"\w{5}", "Hello Python World"))
````

### 🔥 Related Quantifiers (IMPORTANT)

| Quantifier | Meaning               |
| ---------- | --------------------- |
| `{n}`      | Exactly n times       |
| `{n,}`     | At least n times      |
| `{n,m}`    | Between n and m times |


**🔹 {n,} → At Least n Times**

````python
print(re.findall(r"\d{2,}", "Age 9, Code 12345"))
````

**🔹 {n,m} → Between n and m Times**

````python
print(re.findall(r"\d{2,4}", "Years 1 12 123 12345"))
````

**🧠 With Groups (Very Powerful)**
- (ab){3} → ab repeated 3 times

````python
print(re.findall(r"(ab){3}", "ababab ab ababab"))
````

````python
print(re.findall(r"(a){2,3}", "ababab aab abaaabab"))
````

**🧪 Real-World Use Cases**

| Pattern          | Use             |
| ---------------- | --------------- |
| `\d{6}`          | OTP             |
| `\d{10}`         | Mobile number   |
| `[A-Z]{2}`       | State code      |
| `[a-zA-Z0-9]{8}` | Password length |


## 🔹 8. \w → Word Character

- 📌 Matches:

  - a–z
  - A–Z
  - 0–9
  - _
- 🧠 Used for identifiers, usernames


Example

````python
print(re.findall("\w", "Hello_123 World!"))
````

````python
print(re.findall("\\w", "Hello_123 World!"))
````

````python
print(re.findall(r"\w", "Hello_123 World!"))
````

````python
print(re.findall(r"\w+", "Hello_123 World!"))
````

## 🔹 9. \s → Whitespace

- 📌 Matches:

    - Space
    - Tab
    - Newline
- 🧠 Used to detect or remove spaces

Example

````python
print(re.findall("\s", "Hi there"))
````

````python

print(re.findall("\\s", "Hi there"))
````

````python
print(re.findall(r"\s", "Hi there"))
````

## 🔹 10. [abc] → Any One of a, b, or c

- 📌 Matches only ONE character from set
- 🧠 Acts like OR

Example

````python
print(re.findall("[abc]", "apple banana cherry"))
````

## 🔹 11. [^abc] → NOT a, b, or c

- 📌 Matches any character except a, b, or c
- 🧠 ^ inside [] means negation

Example

````python
print(re.findall("[^abc]", "abcXYZ"))
````

# Iterator 👍
- It is an object that allows to programmer to traverse to collection or sequence of data without having to store the entire data in the memory.
- An iterator in Python is an object that allows you to traverse elements one by one.
- It follows the iterator protocol, which means:

- - It implements `__iter__()`

- - It implements `__next__()`
- It doesn't store entire data in memory ,it generate one at a time.
- once an iterator exhausted,can not be read.

- An iterator is an object that contains a countable number of values and returns one value at a time using the __next__() method.

## increamenting iterator object
 - we use two function.
 -  Get iterator using iter() : return iterator object.
 -  Get elements using next() :fetch items one at a time.

# 3️⃣ Iterator (Object that does iteration)
- 👉 What is Iterator?
- An iterator is an object that:
- Implements __iter__()
- Implements __next__()
- Returns one value at a time
- 



```python
# defining our own loop
def my_loop(iterable):
    it = iter(iterable)
    while True:
        try:
            element = next(it)
            print(element)
        except StopIteration:
            break   
my_loop(['a','b','c','d'])
print("=======")
my_loop((1,2,3,4,5))
print("=======")
my_loop({'name':'Alice','age':30,'city':'New York'})
print("=======")
my_loop("Hello")
print("=======")
my_loop(range(1,6))
```

**Note :**
- All iterators are iterable.
- all iterables are iterator.

```python
class myRange:
    def __init__(self, start, stop=None, step=1):
        # Handle single argument case: myRange(5)
        if stop is None:
            self.start = 0
            self.stop = start
        else:
            self.start = start
            self.stop = stop

        self.step = step
        self.current = self.start

    def __iter__(self):
        return self

    def __next__(self):
        if self.step > 0 and self.current >= self.stop:
            raise StopIteration
        if self.step < 0 and self.current <= self.stop:
            raise StopIteration

        val = self.current
        self.current += self.step
        return val
```

```python
for i in myRange(5):
    print(i)
```

```python
for i in myRange(2, 8):
    print(i)
```

```python
# create own range with step
class MyRange:
    def __init__(self,num):
        self.num=num
        self.current=1
    def __iter__(self):
        return self
    def __next__(self):
        if self.current>self.num:
            raise StopIteration
        val=self.current
        self.current+=1
        return val
obj = MyRange(5)
for i in obj:
    print(i)
```

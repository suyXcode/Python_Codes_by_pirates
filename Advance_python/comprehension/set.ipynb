# 🔥 Set Comprehension in Python

* Set comprehension is similar to list comprehension, but it creates a set instead of a list.

**📌 Syntax**

```python
{expression for item in iterable if condition}
```

- ⚠️ Uses curly braces {}
- ⚠️ Automatically removes duplicate values

## ✅ 1️⃣ Basic Example

````python
numbers = {i for i in range(5)}
print(numbers)
````

## ✅ 2️⃣ Remove Duplicates Automatically

* 👉 Even though duplicates exist in list, set keeps only unique values.

````python
nums = [1, 2, 2, 3, 3, 4]

unique_nums = {i for i in nums}
print(unique_nums)
````

## ✅ 3️⃣ With Condition (Filtering)

* 👉 Get only even numbers

````python
even = {i for i in range(10) if i % 2 == 0}
print(even)
````

## ✅ 4️⃣ Mathematical Transformation

* 👉 Square values

````python
squares = {i*i for i in range(5)}
print(squares)
````

## ✅ 5️⃣ From String (Unique Characters)

````python
name = "programming"

unique_chars = {ch for ch in name}
print(unique_chars)
````

## 🎯 Important Differences

| Feature    | List Comprehension | Set Comprehension |
| ---------- | ------------------ | ----------------- |
| Brackets   | `[]`               | `{}`              |
| Duplicates | Allowed            | Not Allowed       |
| Order      | Maintained         | Unordered         |


````python
a = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7]

res = {num for num in a if num % 2 == 0}
print(res)
````

````python
res = {num**2 for num in range(1, 6)}
print(res)
````

````python

t1 = (10, "hello", 5, 3-2j, 11, 'java', (1,2,3), 10, 5, 11)

set1 = {
    i[::-1] if type(i)==str else i
    for i in t1
}

print(set1)
````

````python
t1=(10,"hello",5,3-2j,11,'java',[1,2,3],10,5,11)

set1 = [i[::-1] if type(i) == [list,tuple,dict,set ] or type(i)== str else i for i in t1]
print(set1)
````

````python
n=int(input("Enter the num"))
set1 ={i for i in range(1,n+1) if i%2==1}
print(set1)
````

````python
a = "abc defgh efgh abzqst"

op = {
    (word, len(word) if len(word) % 2 != 0 else len(word) // 2)
    for word in a.split()
}

print(op)
````

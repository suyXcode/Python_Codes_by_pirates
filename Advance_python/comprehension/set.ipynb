```python
a = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7]

res = {num for num in a if num % 2 == 0}
print(res)
```

```python
res = {num**2 for num in range(1, 6)}
print(res)
```

```python

t1 = (10, "hello", 5, 3-2j, 11, 'java', (1,2,3), 10, 5, 11)

set1 = {
    i[::-1] if type(i)==str else i
    for i in t1
}

print(set1)
```

```python
t1=(10,"hello",5,3-2j,11,'java',[1,2,3],10,5,11)

set1 = [i[::-1] if type(i) == [list,tuple,dict,set ] or type(i)== str else i for i in t1]
print(set1)
```

```python
n=int(input("Enter the num"))
set1 ={i for i in range(1,n+1) if i%2==1}
print(set1)
```

```python
a = "abc defgh efgh abzqst"

op = {
    (word, len(word) if len(word) % 2 != 0 else len(word) // 2)
    for word in a.split()
}

print(op)
```

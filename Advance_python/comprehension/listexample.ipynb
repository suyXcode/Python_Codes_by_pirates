# 🔥 List Comprehension in Python

- List comprehension is a concise and pythonic way to create lists in a single line of code. It is faster and more readable than traditional for loops in many cases 🚀

**📌 Basic Syntax**
```python
[expression for item in iterable if condition]
```

* expression → What you want to append
* item → Variable
* iterable → Range, list, tuple, etc.
* condition → Optional filter


## 🎯 Why Use List Comprehension?

* ✔ Cleaner code
* ✔ More readable
* ✔ Slightly faster
* ✔ Pythonic style



---

##  1️⃣ Simple Example

**✅ Traditional Way**
```python
squares = []
for i in range(5):
    squares.append(i*i)

print(squares)
```

**✅ Using List Comprehension**
```python
squares = [i*i for i in range(5)]
print(squares)
```

````python
squares = [i*i for i in range(5)]
print(squares)
````

## ✅ 2️⃣ With Condition (Filter)

**👉 Get only even numbers**

````python
even_numbers = [i for i in range(10) if i % 2 == 0]
print(even_numbers)
````

## ✅ 3️⃣ If-Else in List Comprehension

**👉 Replace odd numbers with "Odd"**


````python
result = [i if i % 2 == 0 else "Odd" for i in range(6)]
print(result)
````

## ✅ 4️⃣ Working with Strings

**👉 Convert words to uppercase**

````python
words = ["python", "java", "c++"]
upper_words = [word.upper() for word in words]
print(upper_words)
````

## zip() 👍
- it is used to combine two or more than two attribute into a single attribute with respect to the index position.
- deal with multiple variable of mupltiple collection in a single for loop.


 var= zip(iterable1,iterable2,iterable3,iterable4,...........,iterablen)


 - it combines the values and store in form of tuple.
 - zip() return zip object
 - attributes having shortest length will be consider

````python
l1=[10,20,30,]
l2=[2.33,3.44,5.55,6.66,7.77]

for i,j in zip(l1,l2):
    print((i,j))
````

````python
l1=[10,20,30,40,50]
l2=[2.33,3.44,5.55,6.66,7.77]

for i,j in zip(l1,l2):
    print((i,j))
````

## Dictionary Function 👍

- In this comprehension we convert operation into new dictionary collection.

Syntax

* var = {key:value for var in collection}
* var = {key:value for var in collection if condition}
* var ={key:value if condition else val2 for var in collection}


````python
nums = int(input("Enter number: "))
result = {i: i*i for i in range(1, nums+1) if i%2==0}
print(result)
````

````python
l1=[10,20,30]
l2=["hii",["hello"],"welcome"]

print({i:j for i,j in zip(l1,l2)})
````

````python
n = int(input("Enter number: "))
result = {i: i*i if i % 2 == 0 else i*i*i for i in range(1, n+1)}
print(result)
````

````python
a = ["Texas", "California", "Florida"] # states
b = ["Austin", "Sacramento", "Tallahassee"] # capital

res = {state: capital for state, capital in zip(a, b)}
print(res)
````

````python
res = {num: num**3 for num in range(1, 6)}
print(res)
````

````python
a=[1,2,5]
b=[2,3,8]
res={i:j for i,j in zip(a,b)}
print(res)
````

````python
n = int(input("Enter number: "))
result = {i: i*i if i % 2 != 0 else i for i in range(1, n+1)}
print(result)
````

````python
n = int(input("Enter number: "))
result = {i: i*i if i % 2 != 0 else i*i*i for i in range(1, n+1)}
print(result)
````

## 🔥 Dictionary Comprehension in Python

- Dictionary comprehension is a concise way to create dictionaries in a single line of code 🚀

**📌 Syntax**
```python
{key_expression: value_expression for item in iterable if condition}
```

* key_expression → What will be the key
* value_expression → What will be the value
* condition → Optional filter

## ✅ 1️⃣ Basic Example

- 👉 Create dictionary of numbers and their squares


````python
squares = {i: i*i for i in range(5)}
print(squares)
````

## ✅ 2️⃣ With Condition (Filtering)

- 👉 Only even numbers

````python
even_squares = {i: i*i for i in range(10) if i % 2 == 0}
print(even_squares)
````

````python
#✅ 3️⃣ Modify Existing Dictionary

marks = {"A": 70, "B": 85, "C": 90}

updated_marks = {k: v + 5 for k, v in marks.items()}
print(updated_marks)
````

````python
# ✅ 4️⃣ Swap Keys and Values
data = {"a": 1, "b": 2, "c": 3}

swapped = {v: k for k, v in data.items()}
print(swapped)

#⚠️ Works only if values are unique.
````

````python
# ✅ 5️⃣ Conditional Value Assignment (If–Else)
result = {i: "Even" if i % 2 == 0 else "Odd" for i in range(6)}
print(result)
````

# 🔥 Difference from Set & List Comprehension
| Type       | Syntax         | Output Type |
| ---------- | -------------- | ----------- |
| List       | `[ ]`          | List        |
| Set        | `{ }`          | Set         |
| Dictionary | `{key: value}` | Dictionary  |

## zip() 👍
 it is used to combine two or more than two attribute into a single attribute with respect to the index position.


 var= zip(iterable1,iterable2,iterable3,iterable4,...........,iterablen)


 - it combines the values and store in form of tuple.
 - zip() return zip object
 - attributes having shortest length will be consider

```python
l1=[10,20,30,40,50]
l2=[2.33,3.44,5.55,6.66,7.77]

for i,j in zip(l1,l2):
    print((i,j))
```

## Dictionary Function 👍

Syntax

* var = {key:value for var in collection}
* var = {key:value for var in collection if condition}
* var ={key:value if condition else val2 for var in collection}


```python
nums = int(input("Enter number: "))
result = {i: i*i for i in range(1, nums+1) if i%2==0}
print(result)
```

```python
l1=[10,20,30]
l2=["hii",["hello"],"welcome"]

print({i:j for i,j in zip(l1,l2)})
```

```python
n = int(input("Enter number: "))
result = {i: i*i if i % 2 == 0 else i*i*i for i in range(1, n+1)}
print(result)
```

```python
n=int(input("Enter the num :"))
l=[i for i in range(1,n+1) if str(i)==str(i)[::-1]]
print(l)
```

```python
n = 3
result = [[i, i*i] for i in range(1, n+1)]
print(result)
```

```python
n = 3
result = [[j for j in (i, i*i)] for i in range(1, n+1)]
print(result)
```

```python
s="cat"
l1=[(ch,i) for ch in s for i in range(len(s))]
print(l1)
```

```python

l1=[(i,j) for i in range(1,4) for j in range(1,4)]
print(l1)
```

```python
result = ["Even" if i % 2 == 0 else "Odd" for i in range(1, 6)]
print(result)
```

```python
# 8️⃣ Flatten a Nested List
# ❓ Question
# Convert [[1,2], [3,4], [5,6]] into a single list.
nested = [[1,2],[3,4],[5,6],[7,8]]
flat = [num for sublist in nested for num in sublist]
print(flat)
```

```python

# 9️⃣ List Comprehension with Function
# ❓ Question

# Convert list of strings to uppercase.


names = ["ram","odd","suraj","mohan"]
upper = [name.upper() for name in names]
print(upper)
```

```python
# 🔟 List Comprehension with Multiple Conditions
# ❓ Question

# Get numbers from 1 to 50 divisible by 3 and 5.

print([i for i in range(1,51) if i%3==0 and i % 5 ==0])
```

```python
# 11️⃣ Remove Vowels from a String
# ❓ Question

# Remove all vowels from "education".

print([ch for ch in "education" if ch not in "AEIOUaeiou"])
```

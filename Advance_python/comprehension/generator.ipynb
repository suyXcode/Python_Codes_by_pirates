# Generator

- It(Comprehension) is used to create generator object.
- 

## syntax
```
var = (i for i in iterable)
```

- next() : by using next() we can fetch  value from generator object.
- fetch one value at a time.

````python
s=(i for i in range(1,10))
print(s)
print(type(s))
print(next(s))
print(next(s))
print(next(s))
print(next(s))
print(next(s))
print(next(s))
print(next(s))
print(next(s))
# print(next(s))
# print(next(s))
# print(tuple(s))
print(list(s))
````

````python
s=[i for i in range(1,10)]
print(type(s))
# print(next(s))
# print(next(s))
# print(tuple(s))
````

````python
s=(i for i in range(1,10))
print(next(s))
print("=======")
for i in s:
    print(i)
````

````python
s=(i for i in range(1,10))
print(next(s))
print(next(s))
print("=======")
for i in list(s)[::-3]:
    print(i)
print(list(s))
````

````python
s=(i for i in range(1,10))
print(next(s))
print(next(s))
print("=======")
for i in list(s)[:-3]:
    print(i)
print(list(s))
````

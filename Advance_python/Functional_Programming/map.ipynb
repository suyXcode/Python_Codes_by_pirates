# map()

- It is inbuilt function which is used to traverse through the items and perfrom same task on each items.
- map() is a functional programming utility that applies a function to every item of an iterable (like list, tuple, etc.).
- It is used to process of modify the iterable
- number of input == no of output
- return map object, so we have typecast it.
- Takes 2 arguments
- - function name
- - Iterable
  
**Syntax**
```python
map(function, iterable)

```

- function → function to apply

- iterable → list, tuple, etc.

- Returns → a map object (iterator)

````python
def square(x):
    return x * x

nums = [1, 2, 3, 4]

result = map(square, nums)
print(list(result))
````

````python
nums = [1, 2, 3, 4]

squares = map(lambda x: x*x, nums)

print(list(squares))
````

## ✅ Multiple Iterables with map()

- map() can take multiple iterables.

````python
a = [1, 2, 3]
b = [4, 5, 6]

result = list(map(lambda x, y: x + y, a, b))
print(result)
````

````python
# 🔥 Example: Convert String Numbers to Int

# Very common in competitive programming 👇

nums = ["1", "2", "3", "4"]

result = list(map(int, nums))
print(result)
````

## ⚡ Example: Multiple Conditions in map()

````python
nums = [-2, 0, 5, -7]

result = list(map(lambda x: "Positive" if x > 0 
                               else "Negative" if x < 0 
                               else "Zero", nums))

print(result)
````

## 🔥 map() vs List Comprehension

| Feature            | map()                                             | List Comprehension |
| ------------------ | ------------------------------------------------- | ------------------ |
| Readability        | Medium                                            | High               |
| Speed              | Slightly faster (in CPython for simple functions) | Very fast          |
| Functional style   | Yes                                               | No                 |
| Multiple iterables | Yes                                               | Needs zip()        |


## 🚀 When to Use map()

- ✔ When applying simple transformation
- ✔ When using built-in functions (int, str, abs)
- ✔ When handling multiple iterables
- ✔ In functional programming style


## ❌ When NOT to Use

- If logic is complex → use normal loop or list comprehension.

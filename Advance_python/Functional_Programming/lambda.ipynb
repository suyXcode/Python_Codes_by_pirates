## Anonymous Function :

- Anonymous function is used to define function in sigle line by the help of lambda keyword.
- it is used to perform simple operations.
- It takes n no arguments but gives a single output.
- simple if-else programs can be done by help of anonymous function.
- it will `return` the value .

**Syntax**
````python
var = lambda argument : Expression


var(values)

```



## ✅ 1️⃣ Basic Lambda Syntax (Single Condition)
```
lambda arguments: value_if_true if condition else value_if_false
```

`````python
check = lambda x: "Positive" if x > 0 else "Negative"
print(check(5))
`````

# ✅ 2️⃣ Multiple if-else (Nested Ternary Operator)

When you want multiple conditions, you nest them from left to right.

## 🔹 Syntax Pattern:

```bash
 lambda args: value1 if condition1 else value2 if condition2 else value3
```

**⚠ Important:**

- Python evaluates from left to right
- Last value acts as the default case

`````python
check = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"

print(check(10))   # Positive
print(check(-3))   # Negative
print(check(0))    # Zero
`````

`````python
def check(x):
    if x > 0:
        return "Positive"
    elif x < 0:
        return "Negative"
    else:
        return "Zero"
`````

`````python
grade = lambda marks: "A" if marks >= 90 else \
                      "B" if marks >= 75 else \
                      "C" if marks >= 50 else \
                      "Fail"

print(grade(82))   # B
`````

# ⚡ 3️⃣ Multiple Conditions Using and / or

`````python
age_check = lambda age: "Adult" if age >= 18 and age < 60 else \
                        "Senior" if age >= 60 else \
                        "Minor"

print(age_check(65))  # Senior
`````

**🚨 Important Rules**

- ✔ Lambda supports only one expression
- ❌ You cannot write normal if-elif-else blocks inside lambda
- ✔ Use nested ternary expressions instead

**🧠 Pro Tip (Important for Interviews)**

If logic becomes complex:

**❌ Avoid this:**

```
lambda x: "A" if x>90 else "B" if x>75 else "C" if x>60 else "D" if x>40 else "F"
```
- ✔ Better to use normal function for readability.



## 🎯 Summary

| Case                | Syntax                                  |
| ------------------- | --------------------------------------- |
| Single if-else      | `lambda x: A if cond else B`            |
| Multiple conditions | `lambda x: A if c1 else B if c2 else C` |
| Complex logic       | Use normal function                     |

`````python
vol = lambda a:"Vowel" if a in "AEIOUaeiou" else "Not"

up = lambda a:True if a.isupper() else False

lo = lambda a:True if a.islower() else False

palin = lambda a: True if str(a) == str(a)[::-1] else False

pos = lambda a : True if a>=0 else False
`````

`````python
# Vowel check
print(vol('a'))     # Vowel

# Uppercase check
print(up('A'))      # upper

# Lowercase check
print(lo('b'))      # lower

# Palindrome check
print(palin("hih"))   # Palindrome

# Positive / Negative check
print(pos(-5))      # Negative
`````

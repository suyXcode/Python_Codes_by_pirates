## Anonymous Function :

- Anonymous function is used to define function in sigle line by the help of lambda keyword.
- it is used to perform simple operations.
- It takes n no arguments but gives a single output.
- simple if-else programs can be done by help of anonymous function.
- it will `return` the value .

**Syntax**
````python
var = lambda argument : Expression


var(values)

```



`````python
x = lambda a : a + 10
print(x(5))
`````

`````python
square = lambda a:a*a
cube = lambda a:a*a*a
reverse = lambda a : str(a)[::-1]
`````

`````python
print(square(10))
print(cube(10))
print(reverse(10))
`````

`````python
g2n = lambda a,b : "a is big" if a>b else "b is big"
print(g2n(10,8))
`````

`````python
vol = lambda a:"Vowel" if a in "AEIOUaeiou" else "Not"

up = lambda a:True if a.isupper() else False

lo = lambda a:True if a.islower() else False

palin = lambda a: True if str(a) == str(a)[::-1] else False

pos = lambda a : True if a>=0 else False
`````

`````python
# Vowel check
print(vol('a'))     # Vowel

# Uppercase check
print(up('A'))      # upper

# Lowercase check
print(lo('b'))      # lower

# Palindrome check
print(palin("hih"))   # Palindrome

# Positive / Negative check
print(pos(-5))      # Negative
`````

`````python
g2n = lambda a,b : "a is big" if a>b else "b is big"
print(g2n(10,8))
`````

`````python
l1=[2.3, 404, 500, 202, 2.2]
for i in l1:
    if palin(i):
        print(i)
`````

`````python
# using filter function
print(list(filter(palin,l1)))
`````

# filter():
- Traverse through the items of collection and return specific values.
- filter() is a built-in function used to select elements from an iterable (list, tuple, string, etc.) based on a condition.

- It keeps elements only when the condition returns True.
- The length of output may be equal and less than the length of the input.
- function which is used in filter function must return boolean output.


**Syntax**
```python
filter(function,iterable)
```


- function → returns True or False
- iterable → list, tuple, set, etc.
- Output → filter object hence typecasting is required(convert to list if needed).
  

## 🔹 Why do we use filter()?

We use filter() to:

- Remove unwanted data
- Select required elements
- Write clean & concise code
- Avoid long for loops

````python
nums = [1, 2, 3, 4, 5, 6]

even = list(filter(lambda x: x % 2 == 0, nums))
print(even)
````

````python
digit = lambda x: x.isdigit()

res=filter(digit, ['1234', 'abcd', '5678', 'efgh'])# True
````

````python
print(list(res))  # ['1234', '5678']
````

````python
vowel = lambda x: x in 'aeiouAEIOU'
res2=filter(vowel, 'Hello World')# True
print(list(res2))  # ['e', 'o', 'o']
````

````python
l1=[4,2,1,6]
res3=lambda n:n%2==1
print(list(filter(res3,l1)))
````

````python
palin=lambda n: str(n) == str(n)[::-1]
print(list(filter(palin, [121, 123, 131, 145, 151])))  # [121, 131, 151]
````

````python
from math import factorial as f
l1 = [3,'hii',4,3.44,'bye',5]
print(list(map(f,filter(lambda x:type(x)==int,l1))))
````

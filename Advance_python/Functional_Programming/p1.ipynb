````python
vol = lambda a:"Vowel" if a in "AEIOUaeiou" else "Not"

up = lambda a:True if a.isupper() else False

lo = lambda a:True if a.islower() else False

palin = lambda a: True if str(a) == str(a)[::-1] else False

pos = lambda a : True if a>=0 else False
````

````python
# Vowel check
print(vol('a'))     # Vowel

# Uppercase check
print(up('A'))      # upper

# Lowercase check
print(lo('b'))      # lower

# Palindrome check
print(palin("hih"))   # Palindrome

# Positive / Negative check
print(pos(-5))      # Negative
````

````python
g2n = lambda a,b : "a is big" if a>b else "b is big"
print(g2n(10,8))
````

# map()

- It is inbuilt function which is used to traverse through the items and perfrom same task on each items.
- It is used to process of modify the iterable
- number of input == no of output
- return map object, so we have typecast it.
- Takes 2 arguments
- - function name
- - Iterable
  
**Syntax**
```python
map(function, iterable)

```



````python
nums = [1, 2, 3, 4]

squares = map(lambda x: x*x, nums)

print(list(squares))
````

````python
def pro_10(n):
    return n * 10

print(list(map(pro_10, (9, 8, 7, 6, 5, 4))))
````

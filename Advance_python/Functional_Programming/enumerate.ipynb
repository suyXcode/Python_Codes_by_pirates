## ✅ What is enumerate()?

- enumerate() adds a counter (index) to an iterable and returns it as an iterator of (index, value) pairs.

- It’s mainly used in for-loops when you need both:

    - ✅ index
    - ✅ value

**Syntax**

```python
    enumerate(iterable, start=0)
```

- iterable → list, tuple, string, etc.

- start → optional starting index (default = 0)

**Returns → iterator of tuples → (index, element)**

## 🔥 Basic Example

````python
names = ["Ram", "Shyam", "Mohan"]

for index, name in enumerate(names):
    print(index, name)
````

## ✅ With Custom Starting Index

````python
names = ["Ram", "Shyam", "Mohan"]

for index, name in enumerate(names, start=1):
    print(index, name)
````

- Useful when displaying serial numbers 🔥

## 🔥 Without enumerate() (Old Way ❌)

````python
names = ["Ram", "Shyam", "Mohan"]

for i in range(len(names)):
    print(i, names[i])
````


**🚨 Less readable**\
**🚨 Not Pythonic**

- ✔ Always prefer enumerate().

# ⚡ Using enumerate() with List

````python
nums = [10, 20, 30]

result = list(enumerate(nums))
print(result)
````

## 🔥 With String

````python
word = "Python"

for index, char in enumerate(word):
    print(index, char)
````

## 🚀 Real Interview Example
- Find index of first negative number


````python
nums = [5, 8, -3, 7]

for i, val in enumerate(nums):
    if val < 0:
        print("Index:", i)
        break
````

## 🧠 Internally How It Works

Equivalent to:

```python

index = 0
for item in iterable:
    yield (index, item)
    index += 1
    
```

- It just adds a counter automatically.

## 🔥 enumerate() vs range(len())

| Feature             | enumerate() | range(len())    |
| ------------------- | ----------- | --------------- |
| Readability         | High        | Medium          |
| Pythonic            | Yes         | Not recommended |
| Clean syntax        | Yes         | No              |
| Risk of Index Error | Low         | Higher          |



## 🎯 When to Use

✔ When you need index + value \
✔ Looping with position tracking\
✔ Printing numbered lists\
✔ Updating values using index



## 💡 Bonus Trick

Modify list using enumerate:


````python
nums = [1, 2, 3]

for i, val in enumerate(nums):
    nums[i] = val * 2

print(nums)
````

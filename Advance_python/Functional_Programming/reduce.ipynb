# reduce()
- It is used to reduce our iterable into a single unit/value.
- reduce() present in the `functools` module.
- it uses a mathematical folding techniques.
- it takes two arguments.
- - function → takes two arguments

- - iterable → list, tuple, etc.

- - Output → single value

**Syntax**
```python
from functools import reduce

reduce(function, iterable)
```



## 🔹 Why do we use reduce()?

- We use reduce() when we need:

- 1. Sum of all elements

- 2. Product of all elements

- 3. Maximum / Minimum value

- 4. Combine elements into one result


## 🔹 How reduce() Works

For list [a, b, c, d]
```text
step1 → function(a, b)
step2 → function(result, c)
step3 → function(result, d)
```





````python
from functools import reduce
````

````python
nums = [1, 2, 3, 4]
result = reduce(lambda x, y: x + y, nums)
print(result)
````

````python
nums = [1, 2, 3, 4]
print(reduce(lambda x, y: x * y, nums))
````

````python
nums = [3, 7, 2, 9, 1]
print(reduce(lambda x, y: x if x > y else y, nums))
````

````python
n = 0
fact = reduce(lambda x, y: x * y, range(1, n + 1), 1)
print(fact)
````

````python
words = ["Python", "is", "powerful"]
print(reduce(lambda x, y: x + " " + y, words))
````

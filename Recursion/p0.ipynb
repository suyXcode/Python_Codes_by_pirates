# 🔁 What is Recursion?
- It is a concept of programming where function call itself repeatedly,until it reaches base case(termination condition) to stop the iteration.
- It follows stack data structure .
- by default recursion limit 1000.

- Recursion is a process where a function calls itself directly or indirectly to solve a problem.

**👉 In other words:**

- A recursive function is a function that calls itself until a certain condition (base case) is met.

### ⚙️ Basic Structure of Recursion

def recursive_function():
    if condition:          # Base condition (stop recursion)
        return result
    else:
        # Recursive call (function calling itself)
        recursive_function()



### 🧠 Two Main Parts of Recursion

## 1️⃣ Base Case:

- The condition that stops recursion.

- Without a base case → recursion runs forever → causes infinite recursion error.

## 2️⃣ Recursive Case:

- The part where the function calls itself to reduce the problem size.



**🧩 Example 1: Print numbers from 1 to 5 using recursion**

```python
def print_numbers(n):
    if n == 0:        # base case
        return
    print_numbers(n - 1)  # recursive call
    print(n)

print_numbers(5)
```



## ✅ Explanation:
---
- Function keeps calling itself until n == 0.

- When recursion starts returning, it prints numbers in reverse order of calls.

🧮 Example 2: Factorial of a number (classic recursion example)

Formula:
𝑛! = 𝑛 × (𝑛 − 1)!        and            0! = 1

---


```python
# Factorial with recursion
def factorial(n):
    if n == 0 or n == 1:    # base case
        return 1
    else:
        return n * factorial(n - 1)   # recursive call

print("Factorial of 5 is:", factorial(5))
```

## 🔢 Example 3: Sum of first N natural numbers
---

```python
def sum_n(n):
    if n == 0:
        return 0
    else:
        return n + sum_n(n - 1)

print("Sum of first 5 numbers:", sum_n(5))
```




sum_n(5) = 5 + sum_n(4)
         = 5 + 4 + 3 + 2 + 1 + 0
         = 15


## 🧾 Example 4: Fibonacci Series using Recursion
---

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

for i in range(7):
    print(fibonacci(i), end=" ")
```

## ✅ Explanation:
---

Each number is the sum of the previous two numbers.

### ⚠️ Important Points 👍
---

| Concept            | Description                                                                    |
| ------------------ | ------------------------------------------------------------------------------ |
| **Base Case**      | Condition where recursion stops                                                |
| **Recursive Case** | Function calls itself with smaller input                                       |
| **Stack Memory**   | Each recursive call is stored in the call stack                                |
| **Tail Recursion** | When recursive call is the last operation in a function                        |
| **Limitation**     | Too deep recursion may cause `RecursionError` in Python (default limit ≈ 1000) |


### 🔧 To Avoid Recursion Error

You can increase recursion limit (not recommended for deep recursions):

```python
# to get recurion limit
import sys
print(sys.getrecursionlimit())


# to set recurion limit
import sys
sys.setrecursionlimit(2000)
print(sys.getrecursionlimit())
```

# 🧠 When to Use Recursion

### ✅ Best suited for:

**Problems that can be divided into smaller subproblems of the same type.**

Examples:

- Factorial
- Fibonacci
- Tower of Hanoi
- Tree/Graph Traversal
- Searching (Binary Search)
- Sorting (Quick Sort, Merge Sort)

| Term                   | Meaning                  | Example          |
| ---------------------- | ------------------------ | ---------------- |
| **Recursion**          | Function calling itself  | `factorial(n)`   |
| **Base Case**          | Stop condition           | `if n == 0`      |
| **Recursive Case**     | Repeated call            | `factorial(n-1)` |
| **Infinite Recursion** | Missing base case        | ❌ Avoid          |
| **Stack Overflow**     | Too many recursive calls | `RecursionError` |

```python
def count_digits(n):
    if n == 0:
        return 0
    return 1 + count_digits(n // 10)

def reverse_num(n, rev=0):
    if n == 0:
        return rev
    return reverse_num(n // 10, rev * 10 + n % 10)

def mirror_number(n):
    rev = reverse_num(n)
    return n * (10 ** count_digits(rev)) + rev


print(mirror_number(8213))
```

```python
# 1. Write a recursive function to calculate the sum of first n natural numbers.
def sum_natural(n):
    if n == 0:
        return 0
    else:
        return n + sum_natural(n - 1)
```

```python
# 2. `Write a recursive function to calculate the factorial of a given number.`
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
```

```python
# 3. Write a recursive function to calculate the nth Fibonacci number.
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)
```

```python
# 4. Write a recursive function to find the greatest common divisor (GCD) of two numbers.
def gcd(a, b):
    if b == 0:
        return a
    else:
        return gcd(b, a % b)
```

```python
# 5. Write a recursive function to reverse a string.
def reverse_string(s):
    if s == "":
        return s
    else:
        return reverse_string(s[1:]) + s[0]
```

```python
# 6. Write a recursive function to check if a string is a palindrome.
def is_palindrome(s):
    if len(s) <= 1:
        return True
    else:
        return s[0] == s[-1] and is_palindrome(s[1:-1])
```

```python
# 7. Write a recursive function to find the power of a number.
def power(base, exp):
    if exp == 0:
        return 1
    else:
        return base * power(base, exp - 1)
```

```python
# 8. Write a recursive function to find the greatest common divisor (GCD) of two numbers using Euclidean algorithm.
def gcd_euclidean(a, b):
    if b == 0:
        return a
    else:
        return gcd_euclidean(b, a % b)
```

```python
# 9. Write a recursive function to find the least common multiple (LCM) of two numbers.
def lcm(a, b):
    if a == 0 or b == 0:
        return 0
    else:
        return abs(a * b) // gcd_euclidean(a, b)
```

```python
# 10. Write a recursive function to find the sum of digits of a given number.
def sum_of_digits(n):
    if n == 0:
        return 0
    else:
        return n % 10 + sum_of_digits(n // 10)
```

```python
# 11. product of digits of a given number
def product_of_digits(n):
    if n == 0:
        return 1
    else:
        return (n % 10) * product_of_digits(n // 10)
```

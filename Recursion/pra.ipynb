```python
def mirror(n):
    if n==1:
        return '1'
    return str(n)+mirror(n-1)+str(n)
print(mirror(4))
```

```python
#Palindrome logic in recursion function

def ispalindrome(n):
    a=str(n)
    if len(a)<=1:
        return True
    elif a[0] != a[-1]:
        return False
    return ispalindrome(a[1:-1])
print(ispalindrome(12321))
        
```

```python
def toh(n,s,a,t):
    if n==1:
        print(f"Move disk from {s} to {t}")
        return
    toh(n-1,s,t,a)
    print(f"Move disk from {s} to {t}")
    toh(n-1,a,s,t)

toh(3,"source","helper","Target")
    
```

```python
#WAJP to calculate power by using recursion.
def power(base, exp):
    if exp == 0:
        return 1
    else:
        return base * power(base, exp - 1)
print(power(2, 5))  # Output: 8
```

```python
#WAJP to get digit count of a number by using recursion.

def count_digits(n):
    # if n == 0:
    #     return 0 
    if n<10:
        return 1
    else:
        return 1 + count_digits(n // 10)

print(count_digits(10))  # Output: 5
```

```python
#WAJP to reverse a number by using recursion.
def reverse_number(n, rev=0):
    if n == 0:
        return rev
    else:
        rev = rev * 10 + n % 10
        return reverse_number(n // 10, rev)
print(reverse_number(1234))  # Output: 4321
```

```python
def isPrime(num,i=2):
    if num <= 1:
        return False
    if i*i > num:
        return True
    if num%i==0:
        return False
    return isPrime(num,i+1)
print(isPrime(13))
```

```python
#WAP to check  a palindrome number by using recursion.
def is_palindrome(n, rev=0):
    if n == 0:
        return rev
    else:
        return is_palindrome(n//10, rev*10 + n%10)

num = int(input("Enter a number to check if it's a palindrome: "))
if num == is_palindrome(num):
    print(f"{num} is a palindrome.")
else:
    print(f"{num} is not a palindrome.")
```

```python
# wap to add value in a list
def addval(a):
    if not a:
        return 0
    return a[0]+addval(a[1:])
l=[10,20,30,40,50]
print(addval(l))
    
```

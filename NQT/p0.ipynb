**First one was to find parking cost. Classic if else problem. It said, if vehicle stays for 2 hours then 100 rs / hr is the charge, if it stayed for less than 5 hours then 50 rs per hour (excluding the first 2 hours which had 100 rs/hr), and if vehicle is parked for more than 5 hours then 20 rs per hour.**

```python
def parking_cost(hours):
    if hours <= 2:
        return hours * 100
    
    elif hours <= 5:
        return (2 * 100) + (hours - 2) * 50
    
    else:
        return (2 * 100) + (3 * 50) + (hours - 5) * 20


# Example usage
hours = int(input("Enter parking hours: "))
print("Total cost: ₹", parking_cost(hours))
```

**There was array given of N size, and there was a X variable which was the max weight. We have to find total number of members who can board the air balloon. (Find all valid members who can board the hot air balloon, given the weight must be less than the sum of individual person’s weight), Array has weights of different person.**

```python
def max_members(weights, X):
    weights.sort()   # Step 1: sort
    total = 0
    count = 0

    for w in weights:
        if total + w <= X:
            total += w
            count += 1
        else:
            break

    return count


# Example
weights = [40, 50, 60, 30, 20]
X = 100
print(max_members(weights, X))  # Output: 3
```

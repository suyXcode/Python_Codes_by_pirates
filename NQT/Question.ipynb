
# Process 👍
## 🔹 What is a Process?

**A process is an independent program in execution.**

- 👉 Example: Running Google Chrome, VS Code, or a Python script — each is a separate process.

# ⚖️ Difference Table (Process vs Thread)
**Question 1. What is the difference between a process and a thread?**

- Process is independet unit of execution with its own memory space, while a thread is a smaller unit within a process that shares the same memory and resources.
- Processes are isolated from each other and require more overhead to create and manage.
- Where Threads , being lightweight,are faster to create and switch but need careful synchronization due to shared memory. A crash in one process doesn't affect others, but a faulty thread can impact the whole process.


**⚖️ Difference Table (Process vs Thread)**

| Feature        | Process 🧱             | Thread 🧵               |
| -------------- | ---------------------- | ----------------------- |
| Definition     | Independent program    | Unit inside a process   |
| Memory         | Separate memory space  | Shared memory           |
| Communication  | Slow (IPC)             | Fast (shared memory)    |
| Creation Time  | Slow                   | Fast                    |
| Resource Usage | High                   | Low                     |
| Isolation      | High (safe)            | Low (risky)             |
| Crash Impact   | Does NOT affect others | May crash whole process |

# 🎯 Simple Real-Life Analogy

- 👉 Think of a process as a company 🏢
- 👉Threads are employees 👨‍💻👩‍💻 inside that company.
    - Each company (Process) works independently
    - Employees (threads) share office resources
    - if one employee messes up -> whole team affected
    - If one company fails -> others still run.


# 🚀 Key Interview Line (Remember this!)

    Process = Independent Execution with separate memory.
    Thread = Lightweight execution sharing memory within a process.

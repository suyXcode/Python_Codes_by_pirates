```python
# create a new list containing only palindrome numbers
a = eval(input("Enter the list:"))
out=[]
for i in a:
    if type(i)==int:
        if str(i)[::-1]==str(i):
            out.append(i)
print(out)
```

```python
def pan(l:list):
    out=[]
    for i in l:
        if type(i)==int:
            if str(i)[::-1]==str(i):
                out.append(i)
    return out

print(pan(["hii","MadaM",99,44,33,22,"level"]))
```

```python
# factor of a number
num=int(input("Enter the number:"))
for i in range(1,num+1):
    if num%i==0:
        print(i,end=",")
```

```python
# count zeros in digit
a=int(input("Enter the number:"))
c=0
for i in str(a):
    if i == "0":
        c+=1
print(c)
```

```python
# prime number :
num=int(input("Enter the number:"))

isprime=True
if num<=1:
    isprime=False
if num==2:
    isprime=True
for i in range(2,num):
    if num%i==0:
        isprime=False
        break
print(isprime)
```

```python
# armstrong number
num=int(input("Enter the number:"))
p=len(str(num))
s=0
for i in str(num):
    s+=int(i)**p
if s==num:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
# reverse a string
s=input("Enter the string:")
print(f"Original string: {s}")
rev=""
for i in s:
    rev=i+rev
print(f"Reversed string: {rev}")
```

```python
# reverse the list
l=[1,2,3,4,5]
rev=[]
for i in l:
    rev=[i]+rev
print(rev)
```

```python
l=[1,2,3,4,5]
rev=[]
for i in l:
    rev.insert(0,i)
print(rev)
```

```python
#reverse the list using append()
l=[1,2,3,4,5]
rev=[]
for i in range(len(l)-1,-1,-1):
    rev.append(l[i])
print(rev)
```

```python
def pascal_triangle(n):
    for i in range(n):
        print(" " * (n - i), end="")
        num = 1
        for j in range(i + 1):
            print(num, end=" ")
            num = num * (i - j) // (j + 1)
        print()

pascal_triangle(5)
```

```python
def forwardslash(n):
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if i + j == n + 1:
                print('*', end=' ')
            else:
                print(' ', end=' ')
        print()
forwardslash(7)
```

```python
def backwardslash(n):
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if i == j:
                print('*', end=' ')
            else:
                print(' ', end=' ')
        print()
backwardslash(7)
```

```python
def cross(n):
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if i == j or i + j == n + 1:
                print('*', end=' ')
            else:
                print(' ', end=' ')
        print()
cross(7)
```

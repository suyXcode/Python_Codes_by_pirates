```python
def print_pattern(n):
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if i==1 or i==n or j==1 or j==n:
                print('*', end=' ')
            else:
                print(" ", end=' ')
        print()
        
print_pattern(5)
```

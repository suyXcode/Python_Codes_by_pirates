```python
def forwardslash(n):
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if i + j == n + 1:
                print('*', end=' ')
            else:
                print(' ', end=' ')
        print()
forwardslash(7)
```

```python
def backwardslash(n):
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if i == j:
                print('*', end=' ')
            else:
                print(' ', end=' ')
        print()
backwardslash(7)
```

```python
#Q:8 Pattern Programming

def cross(n):
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if i == j or i + j == n + 1:
                print('*', end=' ')
            else:
                print(' ', end=' ')
        print()
cross(7)
```

```python
# Q:9 a : Pattern Programming
def horizontalMidline(n):
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if i == (n // 2) + 1:
                print('*', end=' ')
            else:
                print(' ', end=' ')
        print()
horizontalMidline(7)
```

```python
# Q:9 Pattern Programming
def addsymbol(n):
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if j == (n // 2) + 1 or i == (n // 2) + 1:
                print('*', end=' ')
            else:
                print(' ', end=' ')
        print()
addsymbol(7)
```

```python
# hallow square pattern
def hallow_square(n):
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if i == 1 or i == n or j == 1 or j == n:
                print('*', end=' ')
            else:
                print(' ', end=' ')
        print()
hallow_square(5)
```

```python
# inverted right triangle pattern
def inverted_right_triangle(n): 
    for i in range(n, 0, -1): 
        for j in range(1, i + 1): 
            print('*', end=' ') 
        print()
inverted_right_triangle(5)
```

```python
# hollow right triangle pattern
def hollow_right_triangle(n):
    for i in range(1, n + 1):
        for j in range(1, i + 1):
            if j == 1 or j == i or i == n:
                print('*', end=' ')
            else:
                print(' ', end=' ')
        print()
hollow_right_triangle(5)
```

```python
def full_pyramid(n):
    for i in range(1, n + 1):
        for j in range(n - i):
            print(" ", end="")
        
        for k in range(1, 2*i):
            print("*", end="")
        print()
   
full_pyramid(5)
```

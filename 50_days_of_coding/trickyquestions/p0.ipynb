## 🧠 Problem recap

Each word in the sentence contains one digit which represents the position of that word in the final sentence.

**Example**

- "is1 Thi0s T3est 2a" → digit tells position
- Thi0s → position 0 → "This"
- is1 → position 1 → "is"
- 2a → position 2 → "a"
- T3est → position 3 → "Test"

So output = "This is a Test"

*Tricky Question Zoho Problem*

```python
def sort_words(s):
    words = s.split()
    ordered = [None] * len(words)
    for word in words:
        for ch in word:
            if ch.isdigit():
                index = int(ch)
                clean_word = word.replace(ch,"")
                ordered[index] = clean_word
                break
    return " ".join(ordered)

test_string = "is1 Thi0s T3est 2a"
print(sort_words(test_string))  
```

# Polymorphism

- poly means many.
- morphism means forms.

- IT is a phenomena where we have one interface with multiple implementation for a single operation .
- In this due to same name the other operations override the first operations.
- it is three types of polymorphism.
- 1. Method Overloading
- 2. Method Overriding
- 3. Operator Overloading





```python
def sum(a,b,c):
    print(a+b+c)

temp=sum     # Mankey Patching

def sum(a,b): # overriding 
    print(a+b)

sum(12,3)
temp(1,2,3)
```

# Method Overloading

- Method Overloading is a feature of polymorphism where multiple methods performs multiple task by staying in same class with same method name but different parameters

```python
class Area:
    def find_area(self,a=None,b=None):
        if a!=None and b!=None:
            print("Rec:",a*b)
        if a!=None:
            print("Sqr:",a*a)
        else:
            print("nothing")

ob1=Area()
ob1.find_area()
ob1.find_area(12)
# ob1.find_area(12,13)
ob1.find_area(10)
```

# Method Overriding


```python
def add(a,b):
    return a+b
def mul(a,b):
    return a*b

def operation(func,a,b):
    return func(a,b)+func(a,b)

print(operation(add,12,3))
```

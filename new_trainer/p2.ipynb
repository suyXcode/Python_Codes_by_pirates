# Exception Handling

```python
a=int(input("Enter the first num:"))
b=int(input("Enter the second one :"))
print(a/b) #ZeroDivisionError: division by zero when b = 0
```

```python
# Specific Exception Handling for ZeroDivisionError.
def sam():
    try:
        a=int(input("Enter the first num:"))
        b=int(input("Enter the second one :"))
        print(a/b)  
    except ZeroDivisionError:
        print("b should not be zero")
    except ValueError:
        print("B should be int type.")
sam()
```

```python
#Generic Sepcific Exception Handling
def sam1(): 
    try:
        a=int(input("Enter the first num:"))
        b=int(input("Enter the second one :"))
        print(a/b)  
    except Exception as e:
        print(e)    
sam1()
```

```python
#Generic Sepcific Exception Handling
def sam2(): 
    try:
        a=int(input("Enter the first:"))
        b=int(input("Enter the second one :"))
        print(a/b)
    except Exception:
        print("Problem Solved.")
    finally:
        print("Execution Completed")
sam2()
```

# With argument and with out return keyword.

```python
text = "programming"

visited = ""

for ch in text:
    if ch not in visited:
        count = 0
        for i in text:
            if i == ch:
                count += 1
        print(ch, "=", count)
        visited += ch
```

```python
text = "programming"

for ch in set(text):
    print(ch, "=", text.count(ch))
```

```python
text = "programming"

count_dict = {}

for ch in text:
    count_dict[ch] = count_dict.get(ch, 0) + 1

for key, value in count_dict.items():
    print(key, "=", value)
```

```python
text = "programming"
def count_characters(text):
    count_dict = {}

    for ch in text:
        if ch in count_dict:
            count_dict[ch] += 1
        else:
            count_dict[ch] = 1

    print(count_dict)
count_characters(text)
```

```python
def isprime(num):
    if num < 2:
        print(False)
    else:
        flag = True
        
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                flag = False
                break
        
        if flag:
            print(True)
        else:
            print(False)

isprime(13)
```

```python
def is_palindrome(num:int):
    text = str(num)
    if text == text[::-1]:
        print(True)
    else:
        print(False)
is_palindrome(12321)
```

```python
def is_palindrome(num:int):
    rev=0
    n=num
    while num>0:
        digit = num % 10
        rev = rev * 10 + digit
        num = num // 10
    if rev==n:
        print(True)
    else:
        print(False)
is_palindrome(12321)
```
